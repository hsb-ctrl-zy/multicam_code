### **데이터 전처리**

1. **이상치**
    - 값이 이상한 데이터
        - 결측치
        - 범위에서 벗어나는 데이터
        - 값이 크게 벗어나는 데이터 (극단치)
    - 분석 모델의 성능을 떨어뜨리거나 결과에 악영향을 미치는 경우가 발생
    - 이상치 데이터는 즉시 처리
        - 데이터 제거
            - 소량의 데이터가 이상인 경우 제거 (일반적으로 전체 데이터에서 5% 미만)
            - 학습 데이터가 사라진다는 문제가 발생 (다량의 데이터가 이상치로 판단되면 제거 대신 대체)
        - 특정 값으로 대체
            - 일반적으로 평균이나 최빈값으로 대체
            - 데이터의 신뢰도가 떨어질 수 있다.
            - 분석의 성능이 떨어지는 경우 발생 (데이터의 분포가 과밀화되는 경우)
    - 결측치 확인
        - `isna()` 함수를 이용해서 결측치의 유무를 판단 → `sum()` 함수를 이용해서 결측치의 개수를 확인
        - `isna()` 함수를 이용하고 → `any()` 함수를 활용하면 결측치가 포함되어있는 행이나 열을 확인
    - 특정 범위에서 벗어나는 데이터 확인
        - 특정 열에 데이터의 유일값을 출력: `unique()`
        - 특정 열에 데이터의 빈도를 확인: `value_counts()`
        - Series 데이터를 집합의 형태로 변경: `set()`
    - 극단치를 확인하는 방법
        - IQR 방식
            - `boxplot()` 수염의 경계
            - IQR = Q3 - Q1
            - 극단치의 경계
                - 상단: Q3 + 1.5IQR
                - 하단: Q1 - 1.5IQR

<br>

2. **데이터의 불균형 문제**
    - 정상을 정확하게 분류하는 것과 이상을 정확하게 분류하는 것중 **이상을 정확하게 분류하는 것이 중요**
    - 일반적으로 데이터들은 정상이 이상보다 많다.
    - 해당 학습 데이터를 모델에 입력하게 되면 다수의 데이터를 더 학습하여 다수의 라벨로 예측할 확률이 올라간다.
    - 다수의 데이터를 소수의 데이터만큼 줄여주는 방법 ( RandomUnderSampling )
        - 데이터가 제거되므로 학습 데이터가 줄어든다. → 모델의 성능에 영향
        - 제거하여도 데이터의 개수가 충분히 많은 경우
        - 데이터의 신뢰도가 굉장히 중요한 경우
    - 소수의 데이터를 다수의 데이터만큼 늘리는 방법
        - 단순 복제 ( RandomOverSampling )
            - 소수의 데이터를 단순 복제하여 다수의 데이터 개수에 맞춰가는 방법
            - 장점: 데이터의 손실이 없기 때문에 학습 데이터 증가
            - 단점: 과적합 위험성 존재
        - 최근접 데이터 생성 ( SMOTE )
            - 소수 데이터의 관측값에 대한 K개의 최근접 양수를 이웃으로 찾고 관측값과 이웃으로 선택된 값 사이에 임의의 데이터를 생성하는 방법
            - `RandomOverSampling`에서 과적합의 위험을 보완한 샘플링 기법
    - 분류 모델에서 `class_weight` 매개변수를 이용해서 소수의 데이터에 가중치 패널티를 부여하는 방법 존재

<br>

3. **스플릿** (데이터의 분할)
    - 분석 모델에서 학습 데이터와 검증 데이터로 분할
    - 학습시킨 모델의 성능을 평가하기 위함
    - train, test set으로 데이터 분할 (8:2, 7:3)
    - `shuffle` 매개변수 → 분할은 처음부터 80% 까지 train, 나머지는 test로 사용하는 구조에서 데이터 분할 전 데이터들의 순서를 섞어주는가?
        - `shuffle`을 사용하면 안 되는 경우: 시계열 데이터
    - `stratify` 매개변수 → 특정 열을 선택하면 해당 열의 비율에 맞춰서 데이터를 분할 (그룹화)

<br>

4. **스케일링**
    - column 간의 데이터 범위가 크게 차이나는 경우 모델이 정상적으로 작동하지 않는다.
        - 범위가 큰 column이 예측에 상대적으로 큰 영향을 준다.
    - 각 column 간의 범위를 맞춰주기 위한 작업
    - `StandardScaler`
        - 평균이 0, 분산이 1인 정규분포로 스케일링
        - 이상치 민감
    - `MinMaxScaler`
        - 최솟값을 0, 최댓값을 1로 스케일링
        - 이상치 민감
    - `MaxAbsScaler`
        - 최대 절대값과 0이 각각 1, 0이 되도록 스케일링
        - 데이터가 모두 양수인 경우에는 `MinMaxScaler`와 동일하다.
        - 이상치 민감
    - `RobustScaler`
        - 중앙값과 사분위수를 활용하는 스케일링
        - 중앙값을 0으로 설정하고 IQR을 사용하여 이상치의 영향을 최소화

- 불순도: 지니계수

## **· 비지도 학습**
    
#### **차원 축소**

- **PCA**, **t-SNE**
- feature의 개수가 너무 많은 경우 시간을 줄이기 위해서 사용
- 다차원의 데이터를 시각화 시 한계점 발생. 이를 해소하기 위해 사용
- 텍스트데이터를 토큰화한 뒤 단어 간의 유사성을 보여주기 위해서 사용


#### **군집화**

- 비슷한 유형의 데이터들을 모아서 하나의 그룹으로 생성
- **K-Means**: 군집의 개수를 지정할 수 있음
    - 중심점을 랜덤하게 생성하고 주변 데이터들의 위치로 이동하면서 군집 생성
- **DBSCAN**: 군집의 개수를 모델이 정해줌
    - 밀도 기반의 군집화
    - 고밀도 지역의 데이터들을 하나의 군집으로 생성
    - 경계의 데이터들이 군집을 형성할 수 있는가 확인
    - 형성이 가능하면 군집의 크기를 늘려준다.
    - 군집에 포함되어있지 않은 데이터는 노이즈로 판단

<br>
<br>

### **지도 학습 모델의 종류**

1. 선형 회귀 모델 **LinearRegression**
    - 선형 모델의 기본
    - 독립 변수 feature들 간의 선형 함수를 생성하여 결과를 예측하는 모델
    - 비선형 데이터를 예측하면 성능이 좋지 않다.
    - 비선형 데이터를 예측하기 위해서 다항 회귀 (PolynomialFeatures)
        - 제곱항, 상호관계의 feature, 절편항 등으로 feature의 개수를 늘려준다.

<br>

2. 가중치에 패널티를 줌으로써 예측도를 올려주는 모델  
    2-1. 릿지 **Ridge**  
        - L2 패널티를 이용하여 가중치를 0에 가깝게 변화, 과적합을 방지  
        - 과소적합의 위험 증가

    2-2. 라쏘 **Lasso**  
        - L1 패널티를 이용하여 가중치를 0으로 변화  
        - column의 개수가 매우 많은 경우, 선택할 수 있도록 도와준다. (feature selection)  
        - 모델 단순화 가능성 존재

    2-3. 엘라스틱넷 **ElasticNet**  
        - 릿지, 라쏘를 절충한 모델  
        - 두 개의 패널티를 혼합하여 모델의 성능을 올린다.  
        - `l1_ratio` 매개변수를 이용하여 두 패널티의 비율을 설정  

<br>

3. 로지스틱 회귀 **LogisticRegression**
    - 이름은 회귀지만 결과는 분류 모델
    - 0과 1 사이의 값(실수)을 예측하여 0.5보다 큰 경우라면 1, 아니라면 0을 출력하는 알고리즘
    - 종속 변수가 범주형인 경우 적용하는 회귀 분석 방식
    - 종속 변수 y를 직접 모델링에 활용하지 않고, y가 특정 범주에 속하는 확률을 모델링
    - 분류형 모델들은 데이터 불균형에 대한 `class_weight` 존재

<br>

4. 서포트벡터머신 **SVM**
    - 데이터를 분리하는 최적의 경계(결정 경계)를 찾고 해당 경계선에서 어느 정도의 마진(여유 공간) 생성
    - 회귀의 경우, 마진 밖의 데이터들을 안으로 들어올 수 있도록 패널티 부여
    - 분류의 경우, 마진 안의 데이터들을 밖으로 내보낼 수 있도록 패널티 부여

<br>

5. 의사결정 나무 **DecisionTree**
    - 나무 구조를 도표하하여 분류와 예측을 수행하는 분석 방법
    - 적절한 규칙(조건식)을 찾아서 데이터를 분할하며 나무를 성장시키는 과정
    - 불순도를 측정하여 낮은 방향으로 자식 가지를 형성
    - 나무가 깊어질수록 불순도는 감소하지만 과적합의 위험성이 존재

<br>

6. 앙상블 **Ensemble**
    - 단일 결정 트리의 단점을 극복하고자 여러 ML 모델을 연결하여 더욱 강력한 모델을 생성
    - 대표적으로 **배깅(Bagging)**, **부스팅(Boosting)**이 있다.  
    <br>
    6-1. 배깅 **Bagging**  
        - 대표적인 모델: **랜덤 포레스트 (RandomForest)**  
        - 붓스트랩 자료를 생성 (전체 데이터에서 일부의 데이터를 가져와서 모델에 입력)  
        - 여러 모델들을 각각 독립적으로 학습하고 다수결의 원칙에 따라서 최종 결과물을 만들어준다.  
        - 붓스트랩 자료, column의 선택 등을 하는 이유는? → 각 모델들이 데이터 다양성을 가질 수 있도록 하기 위함  
        - 붓스트랩 자료를 통해 선택 받지 못한 데이터들을 OOB(Out of Bag)라 하는데, 이를 검증에서 사용 가능  
        - 장점: 모델의 불안정성을 줄이고 분산 감소  
        - 단점: 편향은 그대로 유지  
    <br>
    6-2. 부스팅 **Boosting**  
        - 대표적인 모델: XGBoost, LightGBM, CatBoost  
        - 여러 개의 모델들을 직렬화하여 하나의 모델의 결과를 이용하여 오차를 계산하고, 오차를 줄이기 위해 다음 모델을 사용하는 형태  
        - 배깅 모델과 달리 직렬로 구성되어 있기 때문에 시간이 오래 걸린다.  
        - 장점: 편향 감소  
        - 단점: 순차적인 학습을 하기에 시간이 증가, 횟수가 너무 많은 경우 과적합, 매개변수 튜닝이 매우 중요

<br>
<br>

### **교차 검증**

1. **KFold**
    - 데이터를 k개의 동일 크기로 나눠준다(폴드화)
    - k번 반복하여 하나의 폴드는 검증용으로, 나머지 폴드는 학습용으로 사용
    - 모든 폴드는 한번씩 검증으로 사용
    - 데이터의 수가 적은 경우, 모델의 성능을 안정적으로 평가할 수 있는 방법
    - 계층화 Kfold도 존재

<br>

2. **GridSearchCV**
    - 매개변수의 조합을 생성하고 최적인 조합의 값을 찾는 방법
    - 조합별로 모델을 학습하고 검증하여 평가 지표를 통해서 좋은 모델의 매개변수 값의 조합을 확인

<br>

3. **Pipeline**
    - 머신러닝의 학습 과정을 하나의 연속된 흐름으로 묶는 도구
    - 데이터의 전처리 과정, 모델의 학습, 예측과 같은 흐름을 하나로 연결하는 기능
    - 장점: 코드 간결화 및 자동화, 데이터의 누수 방지, 재사용성이 높음
    - 단점: 오류 추적의 문제, 학습 시간의 증가, 순차적인 진행이기에 병렬화가 어려움

<br>
<br>

### **평가 지표**
- **회귀 모델**

    - **MAE**
        - 실제값과 예측값의 차이를 절댓값으로 변환하여 평균을 구한 값
    
    - **MSE**
        - 실제값과 예측값의 차이를 제곱하여 평균을 구한 값
        - 실제값과 예측값의 면적을 의미
    
    - **RMSE**
        - MSE 값에 루트를 씌운 값
        - 오차를 제곱하면 기하급수적으로 증가하는 상황이 발생하기에, 해당 현상을 상쇄하기 위해 사용

    - **MSLE**
        - MSE값에 로그를 씌운 값
        - RMSE와 같이 기하급수적으로 커지는 상황을 방지하기 위해 사용
    
    - **MAPE**
        - MAE를 퍼센트로 표시

    - **R2-Score**
        - 1 - (SSE/SST)
        - SSE: 설명하지 못하고 남은 에러
        - SST: 전체 분산
        - 1에 가까울수록 모델의 성능이 좋다.
        
<br>

- **분류 모델**
    
    - 혼동 행렬 **Confusion Matrix**
        - 분류 모델에서 예측 오류가 얼마인가, 어떠한 유형의 오류가 발생하는가를 나타내는 지표
        - TN: 예측이 부정, 실제가 부정
        - FP: 예측이 긍정, 실제가 부정
        - FN: 예측이 부정, 실제가 긍정
        - TP: 예측이 긍정, 실제가 긍정
    
    - 정확도 **Accuracy**
        - 실제의 데이터에서 예측 데이터가 얼마나 같은가 판단
        - ( TN + TP ) / ( TN + FP + FN + TP )
    
    - 정밀도 **Precision**
        - 긍정으로 예측한 것들 중 실제가 긍정인 비율
        - TP / ( TP + FP )
    
    - 재현율 **Recall**
        - 실제 긍정인 것들 중 예측이 긍정인 경우
        - TP / ( TP + FN )
    
    - **F1-Score**
        - 정밀도와 재현율의 조화 평균
        - 2 * TP / ( 2 * ( TP + FP + FN ) )
