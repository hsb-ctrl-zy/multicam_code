```python
import os
from glob import glob
import pandas as pd
```

```python
# 특정 경로에서 파일의 목록을 가져오는 기능
os.listdir('./review')
```

```python
# glob 라이브러리 사용
# 장점: 파일의 경로와 파일명이 동시에 출력
        # 특정 확장자만 목록으로 불러올 수 있다

json_list = glob('./review/*.json')
```

```python
# json_list를 이용하여 여러 파일들을 하나의 데이터프레임으로 결합
# 비어있는 데이터프레임을 생성
total_df = pd.DataFrame()

for file_path in json_list:
    df = pd.read_json(file_path)
    total_df = pd.concat([total_df, df], axis=0)
total_df.reset_index(drop = True, inplace = True)

total_df.info()
```

```python
aspect_df = pd.DataFrame(sum(total_df['Aspects'], []))
```

```python
aspect_df = aspect_df.map(lambda x: x.strip())
```

```python
# 독립 변수에서 중복된 데이터가 존재하면 제거

aspect_df.drop_duplicates('SentimentText', inplace=True)
```

```python
aspect_df.reset_index(drop=True, inplace=True)
```

```python
# 독립 변수 토큰화 벡터화
from konlpy.tag import Okt
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# 토큰화 과정에서 특정 품사들만 사용
# 명사, 동사, 형용사, 부사

okt = Okt()
allow_pos = ['Noun', 'Verb', 'Adjective', 'Adverb']

def tokenize(text):
    result = []
    for word, pos in okt.pos(text):
        if (pos in allow_pos) & (len(word) >= 2):
            result.append(word)
    return result

tokenize(
    aspect_df.loc[0, 'SentimentText']
)
```

```python
vec = TfidfVectorizer(
    tokenizer = tokenize,
    ngram_range = (1, 2),
    min_df = 3,
    max_df = 0.8,
    max_features = 3000
)
```

```python
X = aspect_df['SentimentText'].values
y1 = aspect_df['Aspect'].values
y2 = aspect_df['SentimentPolarity'].values
```

```python
X_vec = vec.fit_transform(X)
```

```python
X_vec.shape
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le1 = LabelEncoder()
y1_le = le1.fit_transform(y1)
le2 = LabelEncoder()
y2_le = le2.fit_transform(y2)
```

```python
from sklearn.svm import LinearSVC
```

```python
svc1 = LinearSVC(class_weight = 'balanced', random_state = 42)
svc2 = LinearSVC(class_weight = 'balanced', random_state = 42)
```

```python
svc1.fit(X_vec, y1_le)
svc2.fit(X_vec, y2_le)
```

```python
pred_1 = svc1.predict(X_vec[-500:])
pred_2 = svc2.predict(X_vec[-500:])
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(pred_1, y1_le[-500:]))
```

```python
print(classification_report(pred_2, y2_le[-500:]))
```

```python
aspect_df['class'] = aspect_df['Aspect'] + aspect_df['SentimentPolarity'].astype(str)
```

```python
aspect_df['class'].value_counts()
```

```python
y3 = aspect_df['class']
le3 = LabelEncoder()
y3_le = le3.fit_transform(y3)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, GridSearchCV
```

```python
pipe = Pipeline(
    [
        (
            'vec', TfidfVectorizer(
                tokenizer=tokenize,
                lowercase=False
            )
        ),
        (
            'model', LinearSVC()
        )
    ]
)
```

```python
cv = StratifiedKFold(n_splits=5)
```

```python
params = {
    'vec__max_features': [None, 3000],
    'model__C': [1.0, 2.0]
}
```

```python
grid = GridSearchCV(
    estimator=pipe,
    param_grid=params,
    cv=cv,
    verbose=1
)
```

```python
grid.fit(X, y3)
```

```python
print('최적의 파라미터:', grid.best_params_)
print('모델 점수:', grid.score)
```

```python
# 종속 변수를 2차원의 데이터 그대로 활용

X = aspect_df['SentimentText'].values
le = LabelEncoder()
aspect_df['Aspect_le'] = le.fit_transform(aspect_df['Aspect'])
aspect_df['SentimentPolarity'] = aspect_df['SentimentPolarity'].astype(int)
y = aspect_df[['Aspect_le', 'SentimentPolarity']].values
```

```python
vec = TfidfVectorizer(
    tokenizer = tokenize,
    ngram_range = (1, 2),
    min_df = 3,
    max_df = 0.8
)
```

```python
X_vec = vec.fit_transform(X)
```

```python
svc = LinearSVC(random_state = 42, class_weight='balanced')
```

```python
from sklearn.multioutput import MultiOutputClassifier
```

```python
# MultiOutputClassifier는 분류모델에서 종속변수가 2차원 이상인 경우 사용하는 객체

multi_model = MultiOutputClassifier(svc)
```

```python
multi_model.fit(X_vec, y)
```

```python
pipe = Pipeline(
    [  
        ('vec', vec),
        ('model', multi_model)
    ]
)
```

```python
params = {
    'model__estimator__C': [0.8, 1.0]
}
```

```python
from sklearn.model_selection import KFold
cv2 = KFold(n_splits = 5, random_state = 42, shuffle = True)
```

```python
grid = GridSearchCV(
    estimator = pipe,
    cv = cv2,
    param_grid = params,
    scoring = 'accuracy'
)
```

```python
grid.fit(X, y)
```

```python
grid.best_params_
```

```python
# test_df에서 RawText의 데이터들을 문장별로 나눠준다.
raw_list = []
raw_dict = {}

for i in range(len(test_df)):
    raw_list.append(
        kkma.sentences(
            test_df.loc[i, 'RawText']
        )
    )

    raw_dict[i] = test_df.loc[i, 'RawText']
```
