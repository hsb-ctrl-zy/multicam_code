```python
from dotenv import load_dotenv
import pymysql
import os
import pandas as pd
import requests
import json
from pprint import pprint

load_dotenv()
```

xml 로드

```python
url = 'https://apis.data.go.kr/B010003/kamcoRlctRlst/pscd'

params = {
    'serviceKey': os.getenv('service_key'),
    'pageNo' : 1,
    'numOfRows': 100,
    'resultType': 'xml'
}

res = requests.get(url, params=params)
```

```python
res
```

```python
# xml 데이터를 dict 형태로 변환: xmltodict 라이브러리 사용
! pip install xmltodict
```

```python
# 데이터의 타입을 내가 쓰기 편한 형태로 변환 → 파싱

from xmltodict import parse
```

```python
print(res.content)
```

```python
res_data_xml = parse(res.content)
```

```python
pprint(res_data_xml['response']['body']['items']['item'])
```

```python
df3 = pd.DataFrame(res_data_xml['response']['body']['items']['item'])
df3.info()
```

json 로드

```python
params2 = {
    'serviceKey': os.getenv('service_key'),
    'pageNo' : 1,
    'numOfRows': 100,
    'resultType': 'json'
}

res2 = requests.get(url, params=params2)
```

```python
res2
```

```python
res_data_json = json.loads(res2.content)
```

```python
pprint(res_data_json)
```

```python
df2 = pd.DataFrame(res_data_json['body']['items']['item'])
```
