#### class 선언

- 생성자 함수 (초기화 함수)
- class 내부에 함수를 선언하는 과정에서 일반적으로 첫번째 매개변수의 이름은 self로 지정
    - 굳이 self로 지정하지 않아도 되지만 통용적으로 self를 사용함
- self: 자기 자신 (생성되는 class 공간을 의미)
- _ _ init _ _ 함수는 class가 생성이 될 때 자동적으로 한번만 호출이 되는 특수 함수
    - 따로 호출할 필요가 없음
- 생성자 함수는 class에서 필요가 없다면 만들지 않아도 무관
- 생성자 함수를 사용하는 이유:
    - class에서 사용할 변수들을 등록하는 과정

```python
class Class_1():
    def __init__(self, _a, _b):
    # class 내부에서 사용할 변수들을 저장
        self.a = _a
        # self.a → self(자기 자신) + .(~안에) + a(공간의 이름)
        self.b = _b
    
    # ______________ 다른 방법 ________________
    
    # def __init__(self, input_a, input_b):
    #     self.a = input_a
    #     self.b = input_b
    
    # def __init__(self, a, b):
    #     self.a = a
    #     self.b = b
    # ________________________________________

    def add_value(self):
    # class 내부에 저장된 self.a 와 self.b를 더해서 되돌려준다.
        result = self.a + self.b
        return result
    
# _______________________________________________________________________________

# 매개변수는 언더바를 적어서 표현할 수도 있음
# 3가지 방법 중 어떤 것을 사용하든 상관 없음
    # 하지만 위 2가지 방법을 추천
    # 변수의 개수가 많아지면 3번째 방법의 경우 입력 변수와 매개 변수를 헷갈릴 수 있기 때문

# __init__ 함수는 class에서 사용할 변수를 저장하는 기능이기 때문에 return을 만들지 않는다.
    # class 생성 시 자동 실행이기 때문에 return으로 받아오는 데이터를 저장하지 않음
    # return 만들면 error

# 주의: line
    # class 안에 def __init__ 하나
        # def 안에 변수 생성
    # class 안에 def add_value 하나
```

```python
class Calculate:

    # 생성자 함수의 선언
    def __init__(self, _a, _b):
        self.a = _a
        self.b = _b
    
    # class에 저장된 변수를 더해서 되돌려주는 함수
    def add(self):
        result = self.a + self.b
        return result
    
    def sub(self):
        result = self.a - self.b
        return result
    
    def multi(self):
        result = self.a * self.b
        return result
    
    def dev(self):
        result = self.a / self.b
        return result
    
    # 처음에 입력했던 데이터가 변하는 부분은 하나도 존재하지 않는다.
```

```python
# class 생성
cal_1 = Calculate(10, 5)
cal_2 = Calculate(3, 7)
```

### 상속
- 클래스를 생성할 때 다른 클래스의 기능을 가지고 와서 사용하는 방법
    - 오버라이드(override)
        - python에서 기본적으로 지원
        - 부모 클래스의 method(함수)의 이름을 다른 기능으로 변경하는 방법
        - 부모 클래스에 a 함수가 있을 때
            - 자식 클래스에 a 함수를 선언(def)하면
                - 자식 클래스에서는 a 함수가 덮어씌워진다.
                - 부모 클래스의 a 함수는 그대로 유지

    - 오버로딩(overloading)
        - python에서 기본적으로 지원해주지 않음
        - 같은 method(함수)를 여러 개 만드는 방법
        - 같은 함수를 호출하는데 인자의 type이 다른 경우
            - : 다른 연산을 하겠다
            - ex) 숫자의 연산, 문자의 연산, 리스트의 연산

```python
# 상속 / 오버라이드

# 부모 클래스 선언
class A:
    def __init__(self, _name, _age):
        self.name = _name
        self.age = _age
    def user_info(self):
        print(f'유저의 이름은 {self.name}이고, 나이는 {self.age}입니다.')
```

```python
user_1 = A('kim', 30)
```

```python
user_1.user_info()
```

```python
# 자식 클래스를 하나 생성
    # → A클래스의 기능을 상속받는다.

class B(A):
    # B 클래스는 A 클래스의 함수들을 사용할 수 있음
    def __init__(self, _name, _age, _addr):
        # _name, _age는 부모 클래스에서 변수로 저장하는 기능 존재
        # self.name = _name
        # self.age = _age
            # 얘네를 사용해도 되지만 이렇게 하면 가독성도 떨어지고 에러가 나기 쉬움
            # 부모 클래스 안에 생성자 함수를 호출 (방법2)
                # 부모 클래스에 접근하는 함수로 super() 를 사용
        super().__init__(_name, _age)
        self.addr = _addr
    def user(self):
        print(f'이름: {self.name}, 나이: {self.age}, 주소: {self.addr}')
    

# 자식 클래스에서 부모 클래스를 상속받는다.
    # = 기능을 물려받는다.
        # 이 기능에 생성자 함수는 제외되어 있다.
# 생성자 함수의 특징 때문에 발생하는 일
    # 해당 이름의 클래스가 생성이 될 때만 실행
# 부모 클래스에서 사용이 되는 변수들을 모두 사용을 하기 위한 방법
    # 1. 부모 클래스와 같이 자식 클래스에서 변수를 모두 생성
    # 2. 부모 클래스의 생성자 함수를 호출
```

```python
user_2 = B('park', 20, 'seoul')
```

```python
user_2.user()
```

```python
user_2.user_info()

# user_2에는 함수가 user() 사용 가능, user_info() 함수도 사용 가능
```

#### User class 선언
- Bank class를 상속받는다.

- '일을 한다'는 함수를 만든다.
    - 어떤 일을 할 것인가? : 매개변수
    - 해당하는 일을 얼마나 하는가? : 매개변수
    - 일의 종류에 따라 시급이 지정이 되어있고 (시급*시간) 값을
        - add_cost() 함수를 이용해서 통장의 잔액이 증가

- '돈을 쓴다'는 함수를 만든다.
    - 어떤 물건을 구매할 것인가?: 매개변수
    - 구매하는 물건에 따라 Bank에 sub_cost() 함수를 이용해서 잔액을 감소
        - 수정: Bank의 sub_cost() 함수에서 성공하는 경우와 실패하는 경우에 return data를 0과 1로 되돌려준다
    - sub_cost() 함수의 결과 값에 따라서 물건의 구매 성공/실패 여부를 출력

- 오버라이드 기능 
    - Bank 클래스에서 user_info() 함수를 User에서 같은 method를 생성하여 덮어씌워준다.
        - 이름과 나이, 현재 잔액과 구매한 물품 목록을 보여준다.
            - 구매한 물품의 목록은 User 클래스의 생성자 함수에서 변수를 생성

```python
# Bank Class 수정 (원본은 밑에 내려가면 연습문제에 있음)

class Bank():
    total_cost = 0
    user_cnt = 0

    def __init__(self, _name, _birth, _cost = 0): # _log = []):
        self.name = _name
        self.birth = _birth
        self.cost = _cost
        self.log = []
        Bank.total_cost += _cost
        Bank.user_cnt += 1
    
    def add_cost(self, _c):
        self.cost += _c
        dict_data = {
            'type' : '입금',
            'cost' : _c,
            'total_cost': self.cost
        }
        Bank.total_cost += _c
        self.log.append(dict_data)
        print(f'입금이 완료되었습니다. 현재의 잔액은 {self.cost} 원 입니다.')
    
    def sub_cost(self, _c):
        if self.cost >= _c:
            self.cost -= _c
            dict_data = {
            'type' : '출금',
            'cost' : _c,
            'total_cost': self.cost
            }
            Bank.total_cost -= _c
            self.log.append(dict_data)
            print(f'출금이 완료되었습니다. 현재의 잔액은 {self.cost} 원 입니다.')
            return 1        #return 1 추가
        else:
            print('잔액이 부족합니다.')
            return 0        #return 0 추가

    #user_info() 함수를 생성하여 이름과 생년월일, 현재 잔액을 출력하는 함수
    def user_info(self):
        print(f'이름: {self.name}, 나이: {self.birth}, 잔액: {self.cost}')
```

```python
# 일의 종류
    # 함수 안에서 여러 조건식을 이용하여 일의 종류, 시급들을 지정
    # DataBase을 이용하여 데이터를 load
    # key:value의 형태를 이용하여 일의 종류:시급이라는 형태의 데이터를 사용
        # 말할걸말할걸말할걸말할걸 굉굉굉...

# 구매할 수 있는 물품 리스트
    # dict 형태의 key:value의 형태를 이용하여 지정이 가능
    # 함수 안에서 여러 조건식 사용 가능
    # DB 데이터를 이용하여 사용 가능

# 두 개의 데이터셋은 dict 형태로 class 변수로 지정
    # 같은 class를 생성했을 때 일의 종류와 물건의 종류는 서로 공유가 가능하게 하기 위해
```

```python
# from typing import override

class User(Bank):

    # 일의 종류와 시급 dict 데이터를 class 변수로 생성
    work_type ={
        'A': 11000,
        'B': 15000,
        'C': 20000
    }

    # 구매하고 싶은 물건
    item_list = {
        '텀블러' : 50000,
        '스위치' : 730000,
        '거치대' : 20000,
        '헤드셋' : 520000,
        '노트북' : 2000000
    }

    # 생성자 함수
    def __init__(self, _name, _birth, _cost):
        # _name, _birth, _cost는 부모 클래스의 생성자 함수를 호출해서 인자로 사용
        super().__init__(_name, _birth, _cost)
        # super().__init__() → 생성이 되는 변수의 개수는?
            # 4개: 이름, 생년월일, 잔액, 거래내역
        # 구매한 아이템의 목록 → 빈 리스트로 생성
        self.items = []

    # 일을 한다 함수 (매개변수: 일의 종류, 시간)
    def work(self, _type, _time):
        # _type이 work_type에 데이터가 존재하는가?
            # dict 데이터에서 key 값들 중에 _type의 값이 포함되어 있는가?
            # in 비교 연산자 → 단일 데이터 in 1차원 데이터
                # → 1차원 데이터 원소 중 단일 데이터가 포함되어 있는가?
        
        if _type in User.work_type:
            # 일의 종류 데이터에 _type이 포함되어 있는 경우
            # 시급 데이터를 load하고 _time과 multiple
            amount = User.work_type[_type] * _time
                # 일당이 계산이 되었으니 일당을 통장(Bank → 부모클래스)에 입금
            super().add_cost(amount)
                # add_cost 라는 함수는 입금완료 출력이 존재
        
        else:
            # print(f'일의 타입이 맞지 않습니다. 현재 잔고는 {self.cost} 원 입니다.')
            # 잘못된 데이터가 들어왔을 때 에러 발생: raise
                # TypeError (데이터의 타입이 맞지 않는다.)
                # ValueError (데이터의 값이 잘못되었다.)
            raise  ValueError('저장되어 있는 일의 종류가 아닙니다.')
                # 이 괄호에서도 f-string 사용 가능

    # 물건을 구매하자 함수 (매개변수: 물건의 종류)
    def buy_item(self, _item):
        # try ~ except 구문
        try:
            # 시도할 코드 작성
            # 물건의 가격을 로드
            # 물건의 가격은 User 클래스 안에 item_list에 dict 형태로 존재
            # key로 사용될 데이터: _item
            # 가격을 알았으니 출금 (Bank 안에 sub_cost())
            # 잔액이 부족하면 0, 정상 출금은 1을 되돌려주기 위해 결과값을 변수에 저장
            sub_res = super().sub_cost(User.item_list[_item])
                #sub_res가 1이면 구매 성공: items에 구매한 물건을 추가
            if sub_res:
                self.items.append(_item)
                print('구매 성공')
            else:
                print('구매 실패')
        
        except Exception as e:
            # try 영역에서 실행되는 코드가 예외가 발생했을 때만 실행
            # except만 쓰면 에러가 나오지 않음
                # 이후 코드들도 실행되기에 유용
                # 하지만 어떤 에러인지 볼 수 없는 단점
            print(type(e).__name__) # 이번 예시에서는 key error 출력
            # 이때 Exception as 를 사용: 에러 메시지를 보여줌
            print(e) # 어떤 부분에서 에러가 났는지 보여줌
            print('물건의 정보가 없습니다.')


    #20260413
    # 오버라이드: 부모 클래스의 기능의 이름을 덮어 씌워서 사용
    # 부모 클래스에서는 1이라는 기능을 하고 자식 클래스에서는 2라는 기능을 한다.
    # 부모 클래스의 함수의 기능은 그대로 유지
    # 이름, 생년월일, 잔액, 구매한 물건의 목록을 출력

    # @override
        # 문법적으로 오버라이드 함수입니다 : 가독성이 좋게 표시
    def user_info(self):
        print(f"""이름: {self.name}, 나이: {self.birth}, 잔액: {self.cost}, 구매한 물건의 목록: {self.items}""")

    # User 클래스에서는 이름 생년월일 잔액 물건의 목록을 출력
    # Bank 클래스에서는 이름 생년월일 잔액을 출력
```

```python
from functools import singledispatch

@singledispatch
def show(data):
    print(f"기본형 함수: {data}")

# show() 함수의 data 매개변수에 인자의 타입이 만약에 int라면
@show.register(int)
def _(data):
    print(f"정수형 함수: {data}")

@show.register(list)
def _(data):
    print(f"리스트형 함수: {data}")
```

```python
show('a')
show(10)
show([1, 2, 3, 4])
```

```python
# 조건문으로 똑같이 만들기
def show2(data):
    if type(data) == str:
        print(f"기본형 함수: {data}")
    elif type(data) == int:
        print(f"정수형 함수: {data}")
    elif type(data) == list:
        print(f"리스트형 함수: {data}")

show2('a')
show2(10)
show2([1, 2, 3, 4])
```

    - 클래스의 이름은 Bank

            - _name: 기본값 X

                - 통장 주인의 생년월일
                - self.birth

                - 금액(_c): 기본값 X
            - class 안에 저장된 self.cost를 _c만큼 증가 (누적합)
            - log에 dict 형태로 {'type': '입금', 'cost': _c, 'total_cost': self.cost}
        - 출금 함수: sub_cost()
            - 매개변수(1개)
                - 금액(_c): 기본값 X
            - self.cost보다 _c가 작거나 같은 경우에만 self.cost를 감소
            - _c가 더 큰 경우에는 self.cost를 유지하면서 '잔액이 부족합니다.'를 출력
            - self.log에 출금이 성공했을 때만 {'type': '출금', 'cost': _c, 'total_cost': self.cost}
                

```python
class Bank():
    def __init__ (self, _name, _birth, _cost = 0, _log = []):
        self.name = _name
        self.birth = _birth
        self.cost = _cost
        self.log = _log
    
    def add_cost(self, _c):
        self.cost += _c
        self.log.append({'type': '입금', 'cost': _c, 'total_cost': self.cost})
    
    def sub_cost(self, _c):
        if _c <= self.cost:
            self.cost -= _c
            self.log.append({'type': '출금', 'cost': _c, 'total_cost': self.cost})
        else:
            print('잔액이 부족합니다.')
```
