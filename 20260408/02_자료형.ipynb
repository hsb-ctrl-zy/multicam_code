### 강의 내용

여러 개의 데이터를 하나의 변수에 대입하기 위한 형태

- tuple
    - 자료형 데이터 중에 가장 기본적인 형태
        : 가장 메모리를 덜 먹는 데이터의 형태
            (강사님 N이죠)
    - 소괄호()를 이용해서 데이터를 묶어서 표현(생성)
    - 한번 생성이 되면 수정이 불가능
        - 원소를 수정하거나 추가 삭제 작업을 할 때는 새로운 tuple을 생성

- list
    - 대괄호[]를 이용해서 데이터를 묶어서 표현(생성)
    - tuple과는 다르게 각 원소들을 수정하거나 추가, 삭제 가능
        - 리스트에서 제공하는 내장 함수를 이용
    - 원소의 개수는 0개부터 생성 가능
        - 비어 있는 리스트 생성 가능
        - 여러 개의 데이터를 하나의 리스트에 추가하여 새로운 데이터의 집합 생성

- dict
    - 중괄호{}를 이용해서 데이터를 묶어서 표현(생성)
    - key: value의 형태로 데이터가 구성 (list에서 index를 key 표현한 방식)
    - key를 기준으로 데이터를 선택, 수정, 삭제 등이 가능
    - 원소의 개수는 0개부터 생성 가능

#### tuple

```python
# tuple 형태
# 원소의 타입이 다르더라도 생성이 가능

tuple_a = (10, 'hello')
tuple_b = (3.14, True)
print(tuple_a, tuple_b)
print(type(tuple_a))
```

```python
# data → 4가지 기능
    #CRUD: Create, Read, Update, Delete
    #tuple의 경우 2가지가 가능

# tuple에서 각각의 원소에 접근(조회)
    # tuple의 원소들은 index(위치)를 기준으로 추출

print(tuple_a[0])
print(tuple_b[1])
```

```python
# tuple 끼리의 합 (문자열끼리의 합과 동일)
print(tuple_a+tuple_b)

# tuple에 숫자를 곱한다면?
print(tuple_a*3)

# tuple과 숫자의 합은 error
```

```python
# 특정 영역의 데이터를 추출

tuple_c = (10, 20, 30, 40)
print(tuple_c[1:3])

# 시작/종료 index의 값을 적지 않는다면 처음부터/마지막까지라고 인식

print(tuple_c[1:])
print(tuple_c[:2])

# 마지막에서 2번째부터 마지막까지

print(tuple_c[-2:])
    #print(tuple_c[**len(tuple_c)**-2:])
        # **해당 부분**이 숨겨져 있는 것

print('-'*16)

# tuple은 문자열이랑 흡사한 부분이 많다.
    # index를 문자열에서 가능할까?
text_a = 'Hello World'
print(text_a[0])
print(text_a[3:7])
print(text_a[5:])
print(text_a[:4])
print(text_a[-2:])
```

#### list

```python
list_a = [30, 20, 40, 10]
list_b = ['abc', 'ad', 'fi', 'be']
```

```python
# 순서가 바뀐다? → 특정 위치의 데이터들을 서로 다른 위치에 대입
# 오름차순 정렬로 변경을 하는 list의 내장 함수: sort()
# list.sort() → 대입 연산자 X → 데이터의 변화 X
    # 인데 얘는 아니다?

print(list_a.sort()) # 확인차 출력이 되지 않는다(=데이터가 변했다)
    # list에 내장된 sort() 함수는 list 자체를 변경
    # list_a의 데이터들을 오름차순 정렬하고 list_a를 변경하는 것까지 함수의 역할
print(list_a)
```

```python
# 순서를 역순으로 변경하는 함수: reverse()
list_b.reverse()
print(list_b)
```

```python
# 내림차순 정렬
# sort와 reverse를 동시에 사용할 수 있을까?

list_a_1 = [10, 203, 34, 239, 458]
list_a_1.sort().reverse()
    # list_a.sort(): 결과 None
    # NoneType 안의 reverse라는 내장 함수는 없음
```

```python
del(list_a_1)

# solution: sorted 사용
# sorted() 함수의 결과가 출력이 됨: 저장이 되지 않는다는 뜻
# reverse()를 사용해서 역순으로 변경은 되는데 어디에 저장할지는 모른다.

list_c = [31, 485, 95, 33, 209]
print(list_c)
sorted(list_c).reverse()
print(list_c) #저장이 되지 않아서 그대로 출력
```

```python
# copy() 와 일반 할당의 차이

list_c = list_a
list_d = list_a.copy()
print(list_c)
print(list_a)
print(list_d)

print('-'*20)

# 대입 연산자를 이용해서 list 데이터를 새로운 변수에 할당
    # = 같은 공간의 이름이 2개
# copy() : 리스트의 데이터들을 복사하여 새로운 공간에 할당

# 단순 확인만을 위한 reverse (변경 X)
print(list(reversed(list_a)))
# [시작위치:종료위치:인덱스증가값]
print(list_a[::-1])
print(list_a[::-2])
```

```python
list_a.reverse()

print(list_a)
print(list_c) # list_c = list_a / 할당 연산자를 통해 복사(?)
print(list_d) # list_d = list_a.copy()

# 할당 연산자를 통한 복사는 같은 공간을 공유하는 것이므로 같이 바뀜 (종속)
# copy는 공간을 새로이 만들어주는 것 (독립)
# 할당 연산자는 룸메가 되는 것이고, copy는 옆집을 구해주는 것...?
```

```python
# 문자 데이터의 정렬
list_b.sort()
print(list_b)
```

```python
# 내림차순 정렬: sort() → reverse()
# sort(): 옵션 변경 시 내림차순도 가능
    # (사기 당한 기분입니다...)

list_b.sort(reverse=True)
print(list_b)
```

```python
# 리스트에 데이터를 추가
# append() : 리스트 가장 뒤에 데이터를 추가 → 데이터만 있으면 됨
# insert() : 특정 위치에 데이터를 추가 → 위치와 데이터 모두 필요
    # insert()... 왜 인서트가 생각이 안 나지

list_a.append(55)
print(list_a)

list_a.insert(3, 35)
print(list_a)
```

```python
list_a.append(40)
print((list_a))
```

```python
# 리스트에서 원소를 제거
# pop() : 리스트 가장 뒤에 데이터를 삭제

list_a.pop()
print(list_a)

```

```python
# remove() : 리스트 안에 특정한 데이터를 제거
list_a.remove(40)
print(list_a)

# 찾으면 해당 데이터만 지우고 종료
```

```python
# del : 리스트의 특정 위치의 데이터를 제거

del list_a[2] # list_a의 3번째 항목을 제거
print(list_a)
```

```python
# for 문으로 2번째 값을 제거하는 방법

new_list = []

for i in range(len(list_a)):
    # list_a의 길이만큼 반복 실행
    # i : 리스트 인덱스 값들
    if i == 1:
        continue
        # list_a[1] 항ㅁ고은 new_list에 추가가 되지 않는다.
    new_list.append(list_a[i])

print(new_list)
```
