```python
# while + break
# 1부터 1씩 증가시켜서 누적합이 1500000이 넘어가는 시점은 언제이며 누적합은 몇인가?

# 누적합 → 값들이 더해질 공간이 필요
# while문에서 시작값을 지정

# while 조건식(bool): 반복 횟수가 명확하지 않을 때
# while True: 무한 루프
# 누적합이 1500000이 넘어가는 시점에는 강제 종료

result = 0
i = 1
while True:
    # 1번 위치
    if result >= 1500000:
        break
    result += i
    i += 1
print("1번 위치:", i, result)
#반복 횟수: 1733회
#i가 1733 출력, 하지만 result에는 1733이 더해지지 않음: 오답



result = 0
i = 1
while True:
    result += i
    # 2번 위치
    if result >= 1500000:
        break
    i += 1
print("2번 위치:", i, result)
#반복 횟수: 1732회
#i가 1732일 때 result에 더해진 상태로 끝남: 정답



result = 0
i = 1
while True:
    result += i
    i += 1
    # 3번 위치
    if result >= 1500000:
        break
print("3번 위치:", i, result)
#반복 횟수: 1732회
#i는 1733 출력, 하지만 result에는 1733이 더해지지 않음: 오답
```

```python
# 누적합이 8400을 넘어가는 i의 값과 누적합의 값을 for문으로 확인

a = 0
for i in range(500):
    i += 1
    a += i
    if a >= 8400:
        break
print("나의 풀이: i -",i,"a -",a)


# 강사님 풀이
result = 0

for i in range(1, 200, 1):
    result += i
    if result >= 8400:
        break
print("강사님 풀이: i -",i,"a -",result)
```

```python
# 1부터 100까지의 합: 5050

result = 0
for i in range(1, 101):
    result += i
print(result)
```

```python
# 1부터 100까지의 짝수의 합: continue 사용
result = 0

for i in range(1, 101, 1):
    if i%2 == 1:
        continue
    result += i
print(result)
```

```python
result = 0

for i in range(1, 101, 1):
    i *= 2
    if i > 100:
        break
    result += i
print(result)
```

```python
# 다중 반복문 : 반복문 안에 반복문이 존재 (2차원 데이터에서 종종 활용)

#첫번째 반복문의 시작
for i in range(5):
    #두번째 반복문이 시작 (첫번째 반복문에서 사용하는 변수 외의 변수명을 사용)
        #첫번째 변수가 제거 당하기 때문
    for j in range(5):
        # print("i="+str(i)+" / j="+str(j))
        # f-string: 문자열 사이에 변수의 값을 대입하여 하나의 텍스트 생성
        print(f'i={i} / j={j}')
# 최종 반복 횟수는 25회
```

```python
# 2개의 주사위가 존재
# 2개의 주사위의 합이 5의 배수인 모든 경우를 출력하시오

for i in range(1,7):
    for j in range(1,7):
        if (i+j)%5 == 0:
            print(f'i={i}, j={j}, i+j={i+j}')

# 한번에 여러가지를 수정하려면 Alt를 누른 상태에서 다른 것도 드래그 선택
# 취소하려면 esc
```
