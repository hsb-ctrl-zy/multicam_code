```python
# 산술 연산자: 기본적으로 숫자형 데이터의 연산 (더하기, 빼기, 곱하기, 나누기, 몫, 나머지, 지수 승)
x = 10
y = 3

print("x + y =",x+y)
print("x - y =",x-y)
print("x * y =",x*y)
print("x / y =",x/y)
print("x // y =",x//y)
print("x % y =",x%y)
print("x ** y =",x**y)
```

```python
# 대입 연산자 → 일반적인 산술 연산자는 변수의 데이터가 변하지 않는다.
    # 할당 연산자를 사용하지 않았기 때문

x = x+3
print(x)

x -= 3
print(x)
```

```python
score = int(input('점수를 입력하시오:'))

if score >= 90:
    print(score, "A학점")
elif score >= 80:
    print(score, "B학점")
elif score >= 70:
    print(score, "C학점")
else:
    print(score, "F학점")

print("프로그램 종료")
```
