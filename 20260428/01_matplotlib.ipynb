#### 연습문제

1. 컬럼의 이름들을 변경
2. 파생변수 → 총 술 소비량: 맥주소비량+위스키소비량+와인소비량
3. 파생변수 → 도수: (순수알콜/총술소비량)*100
4. 대륙별 그룹화하여 도수의 평균을 구한다 → 가장 독한 술을 먹는 대륙은 어디인가?

```python
# 1)
df.columns = ['국가', '맥주소비량', '위스키소비량', '와인소비량', '순수알콜', '대륙']
df
```

```python
# 2)

df['총술소비량'] = df['맥주소비량']+df['위스키소비량']+df['와인소비량']
df
```

```python
# 3)

df['도수'] = df['순수알콜']/df['총술소비량']*100
df
```

```python
#4)

df[df['총술소비량'] == 0]
```

```python
df.fillna(0, inplace=True)
```

```python
# 4)

df[['대륙', '도수']].groupby('대륙').mean()

# 가장 독한 술을 먹는 대륙은 AF이다.
```

```python
import seaborn as sns
```

```python
corr = df.iloc[:, 1:5].corr()
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(data=corr,
            annot=True,
            square=True,
            fmt='0.3f',
            annot_kws={'size': 10},
            cbar = True,
            cmap = 'Blues',
            xticklabels=['맥주', '위스키', '와인', '순수알콜'],
            yticklabels=['맥주', '위스키', '와인', '순수알콜']
            )

# annot: 주석 표시 유무
# square: 히트탭 정사각형 표시 유무
# fmt = 주석의 소수점 자리수를 지정
# annot_kws = 주석의 환경 설정 (size: 글자의 크기를 변경)
# cbar = 사이드바 표시여부
```

```python
sns.pairplot(
    df.loc[:, '맥주소비량':'순수알콜'],
    corner = True,              # 중복이 되는 산점도 그래프 제거
    kind = 'reg',               # 추세선 추가
    plot_kws = {
        'line_kws' : {
            'color' : 'red'     # 추세선의 색상을 변경
        }
    }
)

plt.show()
```

```python
# 과연 우리나라는 총 술 소비량이 전세계 몇위인가?

# 총 술 소비량을 기준으로 내림차순 정렬
df.sort_values('총술소비량', ascending=False, inplace=True)
# 인덱스를 초기화하고 기존의 인덱스를 제거
df.reset_index(drop=True, inplace=True)
```

```python
# 순위 → 위치 값에 +1
# 국가의 이름이 'South Korea' 인 데이터를 추출해서 인덱스를 확인

flag = df['국가'] == 'South Korea'
korea_index = df.loc[flag,].index[0]
korea_index += 1
korea_index
```

```python
df.loc[korea_index, '총술소비량']
```

```python
korea_value = df.loc[korea_index, '총술소비량']
```

```python
# 모든 국가의 총술소비량 데이터를 이용해서 막대그래프 생성

x = df['국가']
y = df['총술소비량']

flag = df['국가'] == 'South Korea'
korea_index = df.loc[flag,].index[0]
korea_rank = korea_index + 1
korea_value = df.loc[korea_index, '총술소비량']

plt.figure(figsize=(20, 8))
plt.bar(x,y)
plt.xticks(rotation = 30)

plt.show()
```

```python
x = df.index
y = df['총술소비량']

flag = df['국가'] == 'South Korea'
korea_index = df.loc[flag,].index[0]
korea_rank = korea_index + 1
korea_value = df.loc[korea_index, '총술소비량']

plt.figure(figsize=(20, 8))
bar_graph = plt.bar(x, y, color='gray')

# 한국 위치의 막대를 선택해서 색상 변경
bar_graph[korea_index].set_color('red')

# 그래프에 주석 추가 (화살표로 주석이 가리키는 곳을 표시)
plt.annotate(
    f"South Korea \n rank: {korea_rank} \n 총술소비량: {korea_value}",
    xy = (korea_index, korea_value),
    xytext= (korea_index+10, korea_value+150),
    arrowprops= {
        'facecolor' : 'pink',       # 화살표 안쪽 색상
        'edgecolor' : 'grey',       # 화살표 테두리 색상
        'headwidth' : 20,           # 화살표 머리 너비
        'headlength' : 20,          # 화살표 머리길이
        'width': 10,                # 화살표 꼬리의 너비
        'shrink': 0.1               # 화살표와 xy, xytext 간의 간격 조정
    }
),
fontsize = 15

# annotate 첫번째 인자: 주석이 들어갈 텍스트 지정
# 두번째 인자: 주석이 어느 부분을 설명하는가? (좌표)
# 세번째 인자: 주석 문구의 위치
# 네번째 인자: 주석과 해당 데이터를 화살표로 연결

plt.show()
```

```python
# 도수를 기준으로 그래프 생성, 주석 추가
df.sort_values('도수', ascending=False, inplace=True)
# 인덱스를 초기화하고 기존의 인덱스를 제거
df.reset_index(drop=True, inplace=True)

x = df.index
y = df['도수']

flag2 = df['국가'] == 'South Korea'
korea_index2 = df.loc[flag2,].index[0]
korea_value2 = df.loc[korea_index2, '도수']
korea_rank2 = korea_index2 + 1

plt.figure(figsize=(20, 8))
bar_graph = plt.bar(x, y, color='gray')

# 한국 위치의 막대를 선택해서 색상 변경
bar_graph[korea_index2].set_color('red')

# 그래프에 주석 추가 (화살표로 주석이 가리키는 곳을 표시)
plt.annotate(
    f"South Korea \n rank: {korea_rank2} \n 총술소비량: {korea_value2}",
    xy = (korea_index2, korea_value2),
    xytext= (korea_index2+10, korea_value2+10),
    arrowprops= {
        'facecolor' : 'pink',       # 화살표 안쪽 색상
        'edgecolor' : 'grey',       # 화살표 테두리 색상
        'headwidth' : 20,           # 화살표 머리 너비
        'headlength' : 20,          # 화살표 머리길이
        'width': 10,                # 화살표 꼬리의 너비
        'shrink': 0.1               # 화살표와 xy, xytext 간의 간격 조정
    },
    fontsize=20
)

# annotate 첫번째 인자: 주석이 들어갈 텍스트 지정
# 두번째 인자: 주석이 어느 부분을 설명하는가? (좌표)
# 세번째 인자: 주석 문구의 위치
# 네번째 인자: 주석과 해당 데이터를 화살표로 연결

# plt.savefig()
# fname 매개변수: 파일의 이름을 지정
# dpi 매개변수: 해상도 지정 → 기본값 100 → 숫자가 올라가면 고해상도
# bbox_inches 매개변수: 그래프 주변 여백을 제거 'tight' 여백 제거하고 저장
# transparent 매개변수: 배경을 제거할 것인가? bool type input
# facecolor 매개변수: 배경의 색을 지정 (transparent 값이 False인 경우 사용 가능)
# show 전에 사용해야한다.

plt.show()
```
