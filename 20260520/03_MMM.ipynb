```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('../data2/Alternative.csv', index_col = 0)
df.head(3)
```

```python
X = df.drop('sales', axis = 1)
y = df['sales']
```

```python
# interaction_only → 상호작용항들만 사용할 것인가?
# include_bias → 상수항인 컬럼을 추가할 것인가?

poly = PolynomialFeatures(degree = 2, interaction_only= True, include_bias=False)
# pd.DataFrame(poly.fit_transform(X))
X_poly = poly.fit_transform(X)
```

```python
poly_columns = poly.get_feature_names_out(X.columns)
poly_columns
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_poly, y, test_size = 0.2, random_state = 42
)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = Ridge()
model.fit(X_train, y_train)
pred = model.predict(X_test)
print(round(r2_score(y_test, pred), 2))
```

```python
coef_df = pd.DataFrame(
    {
        'feature_names' : poly_columns,
        'coefficient' : model.coef_
    }
).sort_values('coefficient', ascending = False)

coef_df
```
