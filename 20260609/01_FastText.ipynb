#### **FastText**

- Word2Vec에서 OOV(Out of Voca) 문제를 해결하기 위한 모델
- Word2Vec은 '강아지'와 '강아지들' 두 단어를 다른 단어로 생각
- FastText는 기존의 Word2Vec 학습 방식은 동일하지만 기준이 되는 문자가 다름
    - Word2Vec은 단어를 기준, FastText는 글자를 기준
    - '강아지' → '강', '아', '지', '강아', '아지', '강아지' (n-gram의 방식)
    - subword를 기준으로 문자들을 벡터화
    - Word2Vec 매개변수들을 이용하지만 추가적으로 subword 관련한 매개변수들을 사용
        - 최소 길이의 subword → `min_n` (default 3)
        - 최대 길이의 subword → `max_n` (default 6)
        - `min_n` = 0, `max_n` = 0 으로 설정하면 subword 사용 금지 → Word2Vec 방식 이용
        - `min_n`이 `max_n`보다 작거나 같게 설정
        - `bucket` 매개변수는 단어(subword) 사전의 개수를 제한 (메모리 과부하 방지)

```python
from gensim.models import Word2Vec, FastText
```

```python
sentences = [
    ['이커머스', '데이터', '분석', '진행'],
    ['상품', '리뷰', '감성', '분석', '합니다'],
    ['형태소', '단위', '임베딩', '가능']
]
```

```python
model_wv = Word2Vec(
    sentences = sentences,
    window = 3,
    vector_size = 50,
    min_count = 1,
    sg = 1,
    epochs = 10
)
```

```python
# 상품명을 기준으로 특정 상품을 검색 시 연관된 상품의 목록을 확인
products = {
    'P001': '무선 이어폰 블루투스 노이즈캔슬링 충전케이스',
    'P002': '유선 이어폰 하이파이 금도금 플러그',
    'P003': '게이밍 마우스 RGB 경량 디자인',
    'P004': '무선 마우스 초경량 블루투스 듀얼모드',
    'P005': '헤드폰 노이즈캔슬링 유선'
}
```

```python
# dict 형태 데이터에서 FastText에서 필요한 데이터는 values
data = products.values()
```

```python
from konlpy.tag import Komoran
```

```python
komoran = Komoran()

tokens_komoran = []
tokens_split = []

for name in data:
    tokens_komoran.append(komoran.morphs(name))
    tokens_split.append(name.split())
```

```python
print(tokens_komoran)
print(tokens_split)
```

```python
ft2 = FastText(
    sentences = tokens_split,
    window = 3,
    min_count = 1,
    sg = 1,
    min_n = 2,
    max_n = 4,
    seed = 42
)
```

```python
# 상품명들을 벡터화
# 단어들의 단위 벡터를 구하고 → 평균을 내고 → L2 정규화를 이용하여 거리를 1로 변경

import numpy as np
```

```python
def sent_vector(token, type = None):
    # token = 토큰화된 문장 데이터
    # type = None인 경우는 평균값, 'l2'인 경우는 L2 정규화
    vectors = []
    for word in token:
        vectors.append(ft2.wv[word])
    
    # vectors가 존재하지 않는 경우 (공백, 텍스트가 사라진 경우)
    if not vectors:
        return np.zeros(ft2.vector_size)
    
    # 단위 벡터들의 평균값을 구한다.
    v = np.mean(vectors, axis = 0)

    # type이 'l2'라면
    if type == 'l2':
        # 벡터의 거리로 나눠준다.
        # 벡터의 거리는? → numpy 거리 함수: linalg()
        v = v / (np.linalg.norm(v) + 1e-12)
    
    return v
```

```python
# token_split을 이용하여 벡터화
items_vector = []
for token in tokens_split:
    items_vector.append(
        sent_vector(token)
    )
```

```python
items_vector_l2 = []
for token in tokens_split:
    items_vector_l2.append(
        sent_vector(token, type = 'l2')
    )
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# 코사인 유사도를 이용해서 가장 근접한 상품의 이름을 출력하는 함수

def recommand_by_text(product_id):
    # dict.keys() 특정 id의 위치를 확인
    idx = list(products.keys()).index(product_id)

    # 해당 id 위치의 단위 벡터 데이터와 다른 상품들의 단위 벡터 데이터를 코사인 유사도를 이용하여 계산
    sims = cosine_similarity([items_vector[idx]], items_vector).ravel()
    
    # 내림차순 정렬
    order = sims.argsort()[::-1]

    # 추천 상품명을 담기 위한 빈 리스트 생성
    result = []
    for i in order:
        # 코사인 유사도가 1인 경우는 자기 자신일 때
        # list(products.keys())[i]의 값이 product_id와 같지 않은 경우를 보아야 함
        if list(products.keys())[i] != product_id:
        # 또는 if sims[i] != 1.0:
            result.append(
                [
                    list(products.values())[i], round(sims[i], 4)
                ]
            )
    return result
```

#### 문제
- 특정한 검색어를 입력했을 때 유사한 상품 목록 3개를 추천하는 함수를 생성
- `items_vector_l2` l2 정규화된 문장 벡터를 이용하여 코사인 유사도를 구하고 상위 3개의 추천 상품명을 출력

처참하게 망한 내 풀이^-^

```python
def recommand_head3(search_text):
    result1 = []
    for idx in range(len(tokens_split)):
        if search_text in tokens_split[idx]:
            result1.append(idx)
    sims = cosine_similarity([items_vector_l2[idx]], items_vector_l2).ravel()
    order = sims.argsort()[::-1]

    # 추천 상품명을 담기 위한 빈 리스트 생성
    result = []
    a = 1
    for i in order:
        if sims[i] != 1.0:
            result.append(
                [
                    list(products.values())[i], round(sims[i], 4)
                ]
            )
            a += 1
            if a == 3:
                break
    return result
```

강사님 풀이

```python
def recommand_by_search(text):
    # 검색어를 이용하여 등록된 상품 목록에서 유사한 상품을 추천
    # text를 이용해서 단위 벡터를 생성
    vector = ft2.wv[text]
    v = vector / (np.linalg.norm(vector) + 1e-12)

    sims = cosine_similarity( [v], items_vector_l2 ).ravel()
    order = sims.argsort()[::-1]

    result = []
    for i in order[:3]:
        result.append(
            [
                list(products.values())[i],
                round(sims[i], 4)
            ]
        )
    
    return result
```
