```python
# 특정 종목 5개의 데이터를 크롤링: 005930, 009150, 000660, 080220, 096770
# 크롤링에서 반복문을 이용해서 사이트에 접속할 때는 딜레이 기능을 부여해야 함
import time
```

```python
# 현재 시간 로드
from datetime import datetime
```

```python
print(datetime.today())
print(datetime.now())

print(datetime.now().strftime('%y%m%d'))
now_str = datetime.now().strftime('%y%m%d')
```

```python
code_list = ['005930', '009150', '000660', '080220', '096770']
now_str = datetime.now().strftime('%y%m%d')

for code in code_list:
    base_url = 'https://comp.wisereport.co.kr/company/c1010001.aspx?cmp_cd='
    df = pd.read_html(base_url+code, encoding='cp949')[3]
    df.to_csv(f"./{code} {now_str}.csv")

    # 시간의 딜레이
    time.sleep(1)
```

```python
code_list = []
# 반복문을 생성하는데 반복 횟수는 명확하지 않다.
while True:
    input_code = input('종목 코드 6자리를 입력하시오. (입력값 종료 시 ENTER)')
    if len(input_code) == 6:
        code_list.append(input_code)
    
    # input_code가 존재하지 않는다면
    if not(bool(input_code)):
        break


now_str = datetime.now().strftime('%y%m%d')

for code in code_list:
    base_url = 'https://comp.wisereport.co.kr/company/c1010001.aspx?cmp_cd='
    df = pd.read_html(base_url+code, encoding='cp949')[3]
    df.to_csv(f"./{code} {now_str}.csv")

    # 시간의 딜레이
    time.sleep(1)
```
