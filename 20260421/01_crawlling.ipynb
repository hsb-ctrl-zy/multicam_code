```python
# 문자 형태로 이루어진 html 문서를 BeautifulSoup class를 이용해서 객체 변환
# 문자 데이터에서는 특정 영역의 content를 추출하기 어렵다.
# BeautifulSoup에 내장된 함수를 이용해서 데이터를 추출하기 위해 class 생성 시 데이터를 대입

soup = bs(html_text, 'html.parser')
```

```python
type(soup)
```

- 태그의 이름을 기준으로 검색
    - soup.태그명
        - 해당 html 문서 안에서 해당 태그 중 첫번째 태그를 되돌려준다.
        - 문자열 함수에서 find() 함수와 비슷한 부분
    - soup.태그명.string
        - 첫번째 태그에서 콘텐츠(태그 안에 데이터)를 되돌려준다.
            - ex) `<p>Test</p>` → 'Test'
    - soup.태그명[속성명]
        - 첫번째 태그에서 특정 속성의 값을 되돌려준다.
            - ex) `<a href = "https://www.naver.com">네이버</a>`
            - → `https://www.naver.com`

```python
print(soup.a)               # a 태그: 하이퍼링크
print(soup.a.string)        # content 출력
print(soup.a['href'])       # 속성값 출력
```

- 내장함수
    - find()
        - html 문서 안에서 특정 태그의 첫번째 정보를 출력
        - find(속성명 = 속성값)
            - 태그들 중 특정 속성에 특정한 값을 가진 태그의 첫번째 정보를 출력
        - 함수의 결과값의 타입은 TAG 데이터의 타입으로 되돌려준다.
    - find_all()
        - html 문서 안에서 특정 태그의 모든 정보를 출력
        - limit 매개변수
            - 찾는 태그의 개수를 지정
        - 함수의 결과값의 타입은 ResultSet 타입으로 되돌려준다.
        - find(), find_all() 함수는 Tag 타입에서는 사용이 가능
        - ResultSet에서는 사용이 불가능
    - ResultSet 타입은 Tag 데이터들이 모여있는 리스트의 형태와 흡사

```python
import requests
from bs4 import BeautifulSoup as bs
```

```python
# 네이버 증권 사이트 크롤링
# requests 라이브러리를 이용해서 html 문서를 받는다.
result = requests.get('https://finance.naver.com')

# 응답 데이터를 문자형으로 변환: BeautifulSoup 파싱
soup = bs(result.text, 'html.parser')
```

```python
soup
```

```python
# html 문서 안에서 태그명이 div이고 class 속성의 값이 section_sise_top인 태그를 모두 찾아서 길이를 구하자

len(
    soup.find_all(
        'div',
        attrs = {
            'class' : 'section_sise_top'
        }
    )
)
```

```python
# 1개라면 find_all()이 아닌 find()를 이용해서 data 추출
div_data = soup.find(
    'div',
    attrs = {
        'class': 'section_sise_top'
    }
)
```

```python
# 8개의 테이블을 모두 찾는다.

tables = div_data.find_all(
    'table',
    attrs = {
        'class': 'tbl_home'
    }
)
```

```python
tables[0]

# thead는 테이블의 제목 (column 이름 → 따로 추출하여 columns 사용)
# tbody는 테이블의 데이터 (values → 따로 추출하여 values 사용)
```

```python
# 반복문에서 사용할 변수명을 그대로 이용하여 1차 작업
table_data = tables[0]
```

```python
# columns에서 사용할 데이터들을 추출 → table 데이터에서 thead 태그를 추출
head_data = table_data.find('thead')
```

```python
# 8개의 테이블 중 1개 완성
    # 8번 반복을 하는 반복문을 생성 
    # → 각각의 데이터프레임이 생성 
    # → 각각 다른 전역변수에 저장(globals)

# 전역변수 이름을 반복 실행할 때마다 변경하기 위한 기본값
vari_num = 1

for table_data in tables:
    # thead 추출: columns
    head_data = table_data.find('thead')
    cols = [th.string for th in th_list]

    # tbody 추출: values
    body_data = table_data.find('tbody')
    tr_list = body_data.find_all('tr')

    values = []

    for tr in tr_list:
        row_list = tr.find_all(['th', 'td'])
        value_data = [data.get_text().strip() for data in row_list]
        values.append(value_data)

    globals()[f"df{vari_num}"] = pd.DataFrame(values, columns=cols)
    vari_num += 1
```
