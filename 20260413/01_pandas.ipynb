### 데이터 load

#### pandas 라이브러리에서 외부의 데이터를 로드하는 함수들
- csv, tsv
    - 가장 단순한 형태의 데이터 파일
    - pandas에 내장된 read_csv() 함수를 사용해서 데이터 로드
    - csv는 데이터를 ',' 기준으로 나눠준다.
        - read_csv(경로)를 이용하여 로드
    - tsv는 데이터를 '  '(탭) 기준으로 나눠준다.
        - read_csv() 함수에서 sep 매개변수에 인자값을 '\t'로 설정

- json
    - 많이 사용하는 데이터 파일 (특히 웹에서)
    - python으로 생각하면 dict 형태와 같다.
    - pandas에서 read_json() 함수를 사용해서 데이터 로드

- xml
    - html 구조 언어와 같이 tag를 이용해서 데이터가 구성
        - ex) <data>'abc'</data>
    - pandas에서 read_xml() 함수를 사용해서 데이터 로드

- 전부 DataFrame으로 바꿔준다.
- 위 3개의 read_xxx() 함수들은 공통적으로 encoding 이라는 매개변수가 존재
    - 기본값은 'UTF-8'
    - 데이터 파일을 로드하는데 해당 데이터셋에 한글이 존재하는 경우에 종종 에러가 발생
        - 이때 인자값을 'CP949', 'EUC-KR'로 사용하면 데이터 로드가 가능

    - 엑셀 파일을 로드하는 함수에는 encoding 매개변수가 존재하지 않는다.

- excel(xls, xlsx)
    - 엑셀 데이터 파일
    - pandas에서 read_excel() 함수를 사용해서 데이터 로드
    - 특정 엑셀 파일을 로드할 때 추가적인 라이브러리 설치가 필요한 경우가 존재
        - 특정 라이브러리가 존재하지 않는다면 모듈을 찾을 수 없습니다 라는 에러가 발생
            - 에러 발생시 해당 라이브러리를 설치 (openpyxl)

- 모든 read_xxx() 함수에서 첫번째 인자값은 파일의 경로를 입력

#### 결측치의 처리
1. 결측치 데이터를 제거
    - 결측치가 포함되어있는 column을 제거
        - 해당 column의 data가 중요하지 않은 경우
    - 결측치가 포함되어있는 index를 제거
        - 일반적으로 결측치가 포함되어있는 index의 수가 전체 data의 수보다 5% 미만인 경우
        - 결측치가 포함된 index의 데이터 신뢰도가 떨어지는 경우
    - 결측치를 제거하는 함수
        - 결측치(na) + 버리다/지우다(drop)
        - dropna()
            - 매개변수 axis : 기준이 되는 축 지정
                - default rows(0)
                - 결측치가 포함되어있는 인덱스를 제거
                - 값을 columns나 1로 변경하면 결측치가 포함된 column을 제거
            - 매개변수 inplace
                - default False
                - 원본 데이터(class 내부에 저장된 변수 데이터)를 변경할 것인가 지정

2. 결측치를 특정한 값으로 채워준다
    - 채우다(fill) + 결측치(na)
    - fillna()
        - 인자에 특정 데이터를 입력
            - 결측치가 해당 인자 값으로 치환
    - ffill: front fill
        - 결측치 직전의 데이터로 대체
    - bfill: back fill
        - 결측치 직후의 데이터로 대체

#### DataFrame의 Filter

- df.loc[ index의 조건식, column의 조건식 ]
    - index의 값을 기준으로 filter (index가 같은 경우(True)에만 출력)
    - column의 값을 기준으로 filter (column이 같은 경우(True)에만 출력)
    - 시작 : 종료
        - → 시작값부터 종료값까지 (종료가 포함)

- df.iloc[ index의 위치, column의 위치 ]
    - index의 위치를 기준으로 filter
    - column의 위치를 기준으로 filter
    - 시작 : 종료
        - → 시작 위치부터 종료 위치 전까지 (종료 위치는 미포함)

- 특정 column의 data를 확인
    - df[단일 column]
        - → 조건이 단일이라면 df는 Series로 변환하여 추출
    - df[ [col1, col2, ...] ]
        - → 조건이 복수의 타입 (ex: list, tuple) 이라면 df로 추출

```python
# map()
    # 1차원 데이터를 각각 원소별 변환이 필요할때 사용하는 함수
# map( 함수, 1차원 데이터 )

list_1 = [1, 2, 3, 4]

def func_1(x):
    return x ** 2


list(
    map(
        func_1, list_1
    )
)
```

```python
# map() + lambda 함수
    # 내가 벽 느꼈던 lambda...

# lambda 함수: 이름 없는 함수 (단순한 계산식을 사용할 때)
    # 휘발성 함수: 한번 사용하고 사라짐
    # lambda 매개변수: 매개변수의 연산
    # return 만들 필요도 없음
        # 네 처음 쓰는 사람 저요 헷갈려하는 거 저요 약간 벽느꼈어요

list(
    map(
        lambda x : x ** 2,
        list_1
    )
)
```

```python
# 기존 파이썬 내장 함수인 map() 은 func, list 모두 요구
    # 출력값 map data
# Series 내장 함수로 map()이 있음
    # Series.map()
    # 데이터는 Series 함수가 담고 있으므로, 우리가 적어줄 것은 'func'뿐
    # 출력값도 series (Series 내장 함수이기에)
```

```python
# Series 안에 map 함수 존재

print(flag_null)

flag_null.map(
    # x 라는 매개변수에는 True, False, True가 순차적으로 입력
        # 이 때 x의 data type은 bool
        # bool 에는 not() 사용 가능
    lambda x : not(x)
)

# lambda 함수는 연산이 간단한 경우에만 사용
```
