```python
# dataset에서 우리가 필요한 데이터가 존재하지 않는다면
    # 데이터를 생성

# 일일 확진자: 오늘의 총확진자 수 - 어제의 총확진자 수
    # 파생변수를 생성 → column 하나 추가
    # 오늘의 총 확진자-어제의 총 확진자

corona.loc[1, '총확진자'] - corona.loc[0, '총확진자']
```

```python
# 0부터 시작해서 일일확진자 column 만들기

new_list = []

for i in range(819):
    # i는 0부터 818까지 대입
    result = corona.loc[i+1, '총확진자'] - corona.loc[i, '총확진자']
    new_list.append(result)

new_list
```

```python
# 1부터 시작해서 일일확진자 column 만들기

new_list2 = [0]

for i in range(1, 820, 1):
    result = corona.loc[i, '총확진자'] - corona.loc[i-1, '총확진자']
    new_list2.append(result)

new_list2
```

```python
# 819번 반복하면서 일일 확진자의 데이터는 819개
    # 첫번째 데이터는 일일 확진자를 구할 수 없어서
# 첫번째 데이터 0을 집어넣어주는 방법
    # list.insert(0,0)
    # 리스트+리스트: 연결
        # [0] + new_list
    # 반복 실행하기 전 new_list를 빈 리스트가 아니라 [0] 대입한 뒤 append()

new_list.insert(0, 0)
```

```python
new_list
```

```python
# corona 데이터 프레임에 new_list로 만들어준 일일 확진자를 대입
# DF는 dict과 비슷한 특성을 가지고 있다.
# DF[new_col] = 1차원 데이터(Series, list, tuple) // 단일 데이터
    # 1차원 데이터를 대입하는 경우는 DF의 index와 길이가 같아야 한다.
    # 단일 데이터를 대입하는 경우는 모든 값이 단일 데이터로 고정된다.

corona['일일확진자'] = new_list2
```

```python
# index 이동 함수 (shift)

corona['총사망자'].shift()  #default 1: 1칸 이동
```

```python
# 1차원 데이터 중 list - list → error
# numpy의 ndarry형태(행렬) → 행렬 - 행렬
    # → 같은 위치의 원소끼리의 차를 구해서 행렬로 
# pandas가 설치될 때 numpy가 같이 설치: pandas에서 numpy를 사용한다.
# DataFrame과 Series의 values 값들은 행렬로 표시
    # Series - Series → 행렬의 차로 계산

corona['총사망자'] - corona['총사망자'].shift().fillna(0)

# 54- NaN = NaN 이었던 건데 fillna(0)을 한다면
    # 54 - 0 로 바뀌기에 54 출력
```

```python
(corona['총사망자'] - corona['총사망자'].shift()).fillna(0)

# 총사망자 shift() 를 이용하여 계산을 하면 결측치가 우선 발생
    # 결측치의 type이 float: 해당 데이터의 타입이 float 변환
```

```python
corona['일일사망자'] = (corona['총사망자'] - corona['총사망자'].shift()).fillna(0)
```
