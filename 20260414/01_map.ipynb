#### iloc/loc

```python
# map() 함수

df = pd.DataFrame(
    {
        'product': [' iphone ', ' iphone', 'iphone',
                    'galaxy ', 'galaxy']
    }
)

df
```

```python
# while문으로 같은 작업
df3 = df.copy()
```

```python
# 내 풀이

i = 0

while True:
    #df3.iloc[i, 0] = df3.iloc[i, 0].strip()
    print(df3.iloc[i, 0].strip())
    i += 1
    if i == len(df3):
        break
```

```python
# 다른 방법

i = 0

while i < len(df3):
    print(df3.iloc[i, 0].strip())
    i += 1
```

```python
# 다른 방법2

i = 0

while True:
    try:
        print(df3.iloc[i, 0].strip())
        i += 1
    except:
        break
```

#### map()

```python
# map() : 1차원 데이터에서 각각의 원소들을 함수에 입력하여 데이터를 수정&연산한다.
# map() 함수는 python 기본 제공이 있고 Series 내장함수로도 존재
# Series.map(데이터의 형태를 변화시킨 함수)

def data_strip(x):
    return x.strip()

df3['product'].map(
    data_strip
)

# Series 형태의 결과 출력
```

```python
# map() + lambda 함수

df3['product'].map(
    lambda x : x.strip()
).value_counts()
```

```python
# Series.strip()이 존재하지 않기에 에러 발생
# Series의 문자 객체 함수에 접근하는 방법
    # Series.str.문자열함수

df3['product'].str.strip()

# Type 변천사
    # DataFrame(df3) → Series(df3['product'])
    # → 문자열 method(df3['product'].str)
    # → Series(df3['product'].str.strip())
```

```python
# 결측치가 들어간 문자열

series_1 = pd.Series(
    ['cat', 'dog', np.nan]
)
```

```python
series_1
```

```python
# Series의 데이터들을 모두 대문자로 변환

series_1.str.upper()
```

```python
# map() 함수를 이용하여 문자를 대문자로 변환

series_1.map(
    lambda x : x.upper()
)

# Error: float 형태
# 결측치의 타입 → float
```

```python
# map() 함수에 na_action 매개변수 존재
    # 결측치 데이터에 함수를 적용할 것인가를 지정

series_1.map(
    lambda x : x.upper(), na_action = 'ignore'
)
```

```python
# 일반적으로 요일 데이터를 숫자형으로 표시
# 0: Sunday, 1: Monday, ..., 6: Saturday

week_series = pd.Series([0, 1, 2, 3, 4, 5, 6])
week_series
```

```python
# week_series에 데이터들을 요일(월, 화, 수, ... , 일)로 변경

week_series.replace(
    {
        0: '일',
        1: '월',
        2: '화',
        3: '수',
        4: '목',
        5: '금',
        6: '토'
    }
)
```

```python
def week_change(x):
    # x에는 0부터 6까지의 숫자들이 대입
    if x == 0:
        result = '일'
    elif x == 1:
        result = '월'
    elif x == 2:
        result = '화'
    elif x == 3:
        result = '수'
    elif x == 4:
        result = '목'
    elif x == 5:
        result = '금'
    else:
        result = '토'
    return result

week_series.map(
    week_change
)
```

```python
# 요일은 0~6 숫자로 이루어져 있다
    # 이 값들을 인덱스로 생각하면?
    # ['일', '월', ..., '토']

def week_change2(x):
    week_list = ['일', '월', '화', '수', '목', '금', '토']
    return week_list[x]

week_series.map(
    week_change2
)
```

```python
gender_series = pd.Series(
    ['f', 'm', 'f']
)
```

```python
# f를 female로 변경하고 m은 male로 변경
# if문을 한 줄로 표현?
    # (참인 경우 출력 데이터) if (조건식) else (거짓인 경우 출력 데이터)

gender_series.map(
    lambda x: 'female' if x == 'f' else 'male'
)
```

```python
list_1 = [1,2,3,4,5,6]

new_list = [x for x in list_1 if x % 2 == 0]
print(new_list)
```

```python
new_list2 = []

for i in list_1:
    if i % 2 == 0:
        new_list2.append(i)

new_list2
```

```python
# [ 변수명 for (조건식) ] → 조건식이 참인 경우의 데이터를 리스트에 대입
# [ (조건식) for ] → 모든 리스트의 데이터를 활용하고 조건식에 맞춰서 대입
# 홀수의 데이터는 홀수 // 짝수의 데이터는 짝수

new_list3 = ['홀수' if x % 2 == 1 else '짝수' for x in list_1]
print(new_list3)
```

```python
new_list4 = ['홀수' if x % 2 == 1 else '짝수' for x in list_1 if x > 3]
print(new_list4)
```

```python
gender_series = pd.Series(
    ['f', 'm', 'f']
)

def gender_name(x):
    if x == 'f':
    	gender = 'female'
    else:
    	gender = 'male'
    return gender

gender_series.map(
    gender_name
)
```
