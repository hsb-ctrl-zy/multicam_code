##### Having

```python
# 직종에서 ANALYST을 제외하고 직종별로 그룹화를 하여 SAL의 평균을 구한 뒤
# 평균의 값이 2500 이상인 데이터에서 평균 SAL이 높은 순서대로 출력
# sql


query_17 = """
    SELECT
        JOB,
        AVG(SAL) SAL_AVG
    FROM `emp`
    WHERE JOB != 'ANALYST'
    GROUP BY JOB
    HAVING SAL_AVG >= 2500
    ORDER BY SAL_AVG DESC
    
"""

sql(query_17)
```
