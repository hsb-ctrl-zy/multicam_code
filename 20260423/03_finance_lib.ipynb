### 라이브러리를 이용해서 데이터 수집

- yahoo finance
    - 글로벌 주식 및 회사 데이터를 로드
- 한국 거래소 정보를 가져오는 라이브러리 (pykrx)
    - 한국 회사들에 대한 데이터 로드

- 실시간 데이터는 불가능
    - pykrx는 일별 주식 가격 조회
    - yfinance는 시간별로 조회 가능 (기간이 굉장히 짧음), 실시간은 아님(한 20분~30분 정도의 차이가 존재)

- 실시간 데이터를 로드하려면 증권사 api 사용
    - 증권사 계정이 존재 (개인 계정 가능)
        - 은행 계좌 등록
        - 별도의 api key 발급

```python
import yfinance as yf
```

```python
# 특정 종목의 데이터를 가져온다.

ticker1 = "005930.KS"
samsung = yf.Ticker(ticker1)
```

```python
samsung.history()
```

```python
# history()
    # interval: 주기
    # period: 기간 설정 (1d, 5d, 1mo, 3mo, 6mo, 1y, 2y)

samsung.history(period='5d', interval='1m')
```

```python
samsung.actions
```
