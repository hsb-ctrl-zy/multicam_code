#### **Doc2Vec**

- Word2Vec의 확장판
- 기본적인 매개변수, 속성, 메서드는 Word2Vec과 같다.
- 추가적인 부분이 존재

---

- **매개변수**
    - `dm`
        - default : 1
        - 학습 알고리즘을 선택
        - 1 : PV-DM
        - 0 : PV-DBOW (빠름)
    - `dm_mean`
        - default : 0
        - 문장 벡터의 계산 방식 (PV-DM의 경우에만 사용)
        - 1 : 벡터들의 평균
        - 0 : 벡터들의 합산 → 범위가 커지는 경우 발생 → 노이즈로 인한 성능 저하가 발생할 수 있다.
    - `dm_concat`
        - default : 0
        - PV-DM 에서 문서 벡터와 문장 벡터를 결합할 것인가?
        - 결합하는 경우 차원이 급격하게 증가: 과적합 위험 ↑
    - `dbow_words`
        - default : 0
        - PV-DBOW 사용 시 단어 벡터도 동시에 학습할 것인가?
    - `alpha` / `min_alpha`
        - default : 0.025, 0.0001
        - 보폭의 크기
        - 반복 학습을 할 때 경사하강법의 기본적인 매뉴얼 → 너무 크면 둔감, 너무 작으면 매우 민감


---

- **속성**
    - `model.wv`
        - 학습이 된 단어 벡터의 저장소
        - PV-DBOW 방식을 이용하고 dbow_words가 0인 경우 → 문서 벡터를 기준으로 단어 벡터를 생성 (단어 벡터를 학습으로 지정했을 때랑 비교하면 신뢰도가 떨어진다.)
    - `model.dv`
        - 학습이 된 문장 벡터의 저장소
    - `model.dv.index_to_key`
        - 문서들의 ID값 리스트

---

- **메서드**
    - `build_vocab()`
        - 단어 / 단서 사전을 구성
        - 매개변수 `update` : 기존 사전의 갱신을 할 것인가? (default = False)
    - `train()`
        - 사전이 구축된 뒤 직접 학습을 수행
        - 기존에 학습된 모델에 train() 함수를 사용하게 되면 추가적인 학습 진행
    - `infer_vector()`
        - 새로운 문장을 벡터화: 추가적인 학습이 진행되지 않는다.
        - 학습된 모델을 사용해 문장 벡터를 되돌려준다.

---

- `build_vocab()`, `train()` 메서드를 이용하여 수동으로 학습 설정
    - 기존에 학습된 모델에 파인튜닝(추가 학습)도 가능
- 일반적으로는 객체 생성시 sentences에 데이터를 추가함으로 학습을 실행
- 파인 튜닝을 사용하는 이유?
    - 데이터의 양이 증가하면 성능이 오를 수 있다.
    - 분할 학습을 통해서 시간을 나눠서 사용 가능
    - 기존에 학습시킨 데이터와 유사한 데이터를 파인 튜닝하여 성능을 향상

```python
from konlpy.tag import Komoran
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
from sklearn.metrics.pairwise import cosine_similarity

import numpy as np
import re
```

```python
docs = [
    '#나는 커피를 정말 좋아한다',
    '%오늘 아침에 에스프레소 두 잔을 마셨다',
    '카페라떼가 제일 맛있다고 생각한다',
    '나는 차를 더 자주 마신다',
    '녹차를 마시면 기분이 편안해진다',
    '홍차는 향이 깊고 고급스러운 느낌이다',
    '카페에서 책을 읽는 시간이 너무 좋다',
    '허브티도 몸에 좋은거 같다'
]
```

```python
komoran = Komoran()

allow_pos = ['NNP', 'NNG', 'VV', 'VA', 'MAG', 'SL']
stop_word = ['하다', '되다', '이다', '것', '수', '거']
```

```python
# 문자에서 불필요한 글자들을 제거 (정규화)

def nomalize(text):
    # 특수 문자, 공백에 대한 처리
    text = re.sub(r'[^가-힣0-9a-zA-Z\s\.]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
nomalize_docs = [nomalize(doc) for doc in docs]
nomalize_docs
```

```python
def tokenize(text):
    text = nomalize(text)
    print(text)

    tokens = []
    for word, pos in komoran.pos(text):
        if pos in allow_pos:
            if pos in ['VV', 'VA']:
                word += '다'
            if word not in stop_word and len(word) > 1:
                tokens.append(word)
    return tokens
```

```python
tokenize(docs[0])
```

```python
tokenize_docs = [tokenize(doc) for doc in docs]
tokenize_docs
```

```python
# doc2vec은 문장 하나씩 ID를 붙여준다.
tagged = []
for idx, token in enumerate(tokenize_docs):
    # idx: 위치
    # token: 리스트의 원소
    tagged.append(
        TaggedDocument(words = token, tags = [f'DOC_{idx}'])
    )

tagged
```

```python
model = Doc2Vec(
    documents = tagged,
    vector_size = 100,
    window = 3,
    min_count = 1,
    dm = 1,
    dm_mean = 1,
    epochs = 100,
    seed = 42,
    negative = 5        # 잘못된 단어 배치를 사용하여 학습에서 이용
)
```

```python
len(model.dv)
```

```python
len(model.wv)
```

```python
def print_similarities(doc_id):
    # 입력받은 id 문장의 벡터 값
    dic_vec = model.dv[f'DOC_{doc_id}']
    
    # 각각 문장들의 벡터 데이터를 리스트로 생성
    mat = np.array(
        [model.dv[f'DOC_{i}'] for i in range(len(docs))]
    )

    # 문장 벡터들을 dic_vec와 코사인 유사도를 구한다.
    sims = cosine_similarity([dic_vec], mat)[0]

    order = sims.argsort()[::-1]

    # 기존의 문장 출력
    print(f'기존의 문장: {doc_id} → {docs[doc_id]}')

    for idx in order:
        print(f' → {idx} {docs[idx]}  |  cos = {sims[idx]}')
```

```python
print_similarities(2)
```

```python
# 고차원의 벡터 데이터를 그래프 시각화 하기 위해서 차원 축소 라이브러리 호출

from sklearn.manifold import TSNE
```

```python
X = np.array(
    [model.dv[f'DOC_{idx}'] for idx in range(len(docs))]
)

X.shape
```

```python
tsne = TSNE(
    n_components = 2,
    perplexity = 5,
    random_state = 42,
    max_iter = 2000
)
```

```python
X_2d = tsne.fit_transform(X)
X_2d.shape
```

```python
import matplotlib.pyplot as plt
import platform
```

```python
if platform.system() == 'Windows':
    plt.rc('font', family = 'Malgun Gothic')

# 마이너스 기호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False
```

```python
plt.figure(figsize = (12, 8))

# 산점도 그래프 생성
plt.scatter(
    X_2d[:, 0], X_2d[:, 1]
)

# 각각의 점에 주석
for idx, text in enumerate(docs):
    plt.annotate(
        text,
        xy = (X_2d[idx, 0], X_2d[idx, 1]),
        xytext = (3, 3),
        textcoords = 'offset points'    # 각각의 위치에 텍스트 대입
    )

plt.show()
```

```python
# 문장별 벡터의 값을 이용하여 ML 학습
# 독립변수: 문장 벡터
# 종속변수: 커피(1), 논커피(0) 데이터 생성

y = np.array([1, 1, 1, 0, 0, 0, 1, 0])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.25, random_state = 42, stratify = y
)
```

```python
clf = LogisticRegression(random_state = 42, max_iter = 2000)
```

```python
clf.fit(X_train, y_train)
```

```python
pred = clf.predict(X_test)
print(classification_report(y_test, pred))
```
