#### 문제
- pandas를 이용해서 data 폴더 안 'rating_train.txt' 파일 로드
- 결측치, 중복값 제거
- 상위 500개 데이터 필터링
- document 데이터와 label 데이터로 나눠준다. (독립/종속)
- embed 라이브러리 이용하여 토큰화, 벡터화 작업
- 분류 모델은 svc, logistic 모델 이용 (parameter defalut)
- 벡터화 방법 2가지를 이용하고 모델 2개를 이용하여 성능 평가
- 최적의 모델을 이용하여 하위 200개의 데이터를 예측하고 분류 레포트 생성

```python
import pandas as pd

df = pd.read_csv('../data/ratings_train.txt', sep='\t')
df.head()
```

```python
df.drop('id', axis = 1, inplace = True)
```

```python
df.dropna(inplace=True)
```

```python
df.drop_duplicates('document', inplace=True)
```

```python
X = df['document'][:500].values
y = df['label'][:500].values
```

```python
X
```

```python
y
```

```python
import embed
```

```python
tokens = embed.global_vari_set(X)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import classification_report

logi_wv = LogisticRegression(random_state = 42)
logi_idf = LogisticRegression(random_state = 42)
svc_wv = SVC(random_state = 42)
svc_idf = SVC(random_state = 42)
```

```python
X_wv = []
X_idf = []

for token in tokens:
    vec_wv = embed.sent_embed_mean(token)
    vec_idf = embed.sent_embed_wv_idf(token)
    X_wv.append(vec_wv)
    X_idf.append(vec_idf)
```

```python
print(embed.run_model(X_wv, y, model = logi_wv))
print(embed.run_model(X_idf, y, model = logi_idf))
```

```python
print(embed.run_model(X_wv, y, model = svc_wv))
print(embed.run_model(X_idf, y, model = svc_idf))
```

```python
# 최적의 모델(svc_idf)을 이용하여 예측

X_test = df['document'].tail(100).values
y_test = df['label'].tail(100).values
```

```python
pred = embed.predict_sentence_list(X_test, svc_idf, 'idf')
```

```python
pred_df = pd.DataFrame(pred, columns = ['sentence', 'label'])
pred_df
```

```python
pred_y = pred_df['label'].map({
    '부정': 0,
    '긍정': 1
})
```

```python
pred_y
```

```python
print(classification_report(y_test, pred_y.values))
```
