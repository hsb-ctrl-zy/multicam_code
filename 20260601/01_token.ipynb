#### Okt

- SNS, 댓글 등의 한글로 이루어진 데이터를 형태소 분석하기 위함

<br>

- 주요 품사 태그 종류
    - Noun: 명사
    - Verb: 동사
    - Adjective: 형용사
    - Josa: 조사
    - Adverb: 부사
    - KoreanParticle: 한국어 파티클 / 이모티콘 (ex. ㅋㅋㅋ, ㅠㅠ)
    - Punctuation: 구두점 ( . , ? ! ~ )
    - Number: 숫자
    - Alpha: 영문자
    - Modifier: 관형사 (새, 헌, 첫)
    - Foreign: 외국어 / 한자
    - ScreenName: 트위터 아이디 표현 (@username)
    - Hashtag: 해시태그 (#Tag)

<br>

- 매개변수
    - 정규화 (norm)
        - 문자를 정규화
        - 비표준어, 오타, 불필요한 반복
        - 문자들에서 노이즈를 제거
    - 어간 추출 (stem)
        - 사전에 등록된 기본 형태만 남기는 방법
        - 한국어의 동사와 형용사는 문맥에 따라서 형태가 매우 많다. 꼬리(어미)를 잘라내고, 기본형으로 통일

```python
from konlpy.tag import Okt
```

```python
okt = Okt()
```

```python
text = "멀티캠퍼스 부트캠프의 데이터 분석가 과정을 수강하고 있습니다ㅎㅎ"

tokens = okt.morphs(text)
print("tokens:", tokens)
print("split:", text.split())
```

```python
text2 = "멀티캠퍼스부트캠프의데이터분석가과정을수강하고있습니다ㅎㅎ"

tokens2 = okt.morphs(text2)
print("tokens:", tokens2)
print("split:", text2.split())
```

```python
text = "파이썬에서 자바의 환경 변수 설정은 너무 힘이 듭니다. ㅠㅠ 오프라인 수업이 그립습니닼ㅋㅋ"

# 형태소 분리
tokens = okt.morphs(text)
print(tokens)
```

```python
text.split()
```

```python
text2 = "파이썬에서자바의환경변수설정은너무힘이듭니다.ㅠㅠ오프라인수업이그립습니닼ㅋㅋ"
```

```python
tokens2 = okt.morphs(text2)
print(tokens2)
```

```python
text2.split()
```

```python
# 명사 추출
nouns = okt.nouns(text)
print(nouns)
```

```python
# 품사 태깅
# 단어와 해당 단어의 품사를 튜플의 형태로 묶어서 반환

print(okt.pos(text))
```

```python
# Josa, Punctuation 품사를 제외하고 나머지 문자들만 사용

new_list = []

for t, p in okt.pos(text):
    # t에는 단어가, p에는 품사가
    if p not in ['Josa', 'Punctuation']:
        new_list.append(t)

new_list
```

```python
# 정규화, 어간 추출
# norm: 문자의 오타, 비표준어, 불필요한 반복을 다듬어주는 부분
# stem: 꼬리(어미)를 잘라내고 기본형으로 변환

okt.pos(text, norm = True)
```

```python
# 정규화, 어간 추출

okt.pos(text, norm = True, stem=True)
```

#### **kkma 분석기**
- 굉장히 깐깐한 언어 분석기
- 국어 사전에 나온 단어들을 이용하여 단어를 토큰화하는 분석기

#### **세종 태그셋**
- 명사류
    - `NNG`: 일반 명사
    - `NNP`: 고유 명사
    - `NNB`: 의존 명사
    - `NP`: 대명사
- 동사/형용사
    - `VV`: 동사
    - `VA`: 형용사
- 조사
    - `JKS`: 주격 조사 (이, 가)
    - `JKO`: 목적격 조사 (을, 를)
    - `JKM`: 부사격 조사 (에, 에서, 으로)
- 기타
    - `MAG`: 일반 부사
    - `EFM`: 평서형 종결 어미 (~다, ~는다)
    - `SF`: 마침표, 물음표, 느낌표

```python
# 꼬꼬마
from konlpy.tag import Kkma
```

```python
kkma = Kkma()
```

```python
# 문장 분리용 샘플 텍스트 생성
test_formal = '인공지능 데이터 분석가 과정에 오신 것을 환영합니다 오늘부터 자연어 처리를 배우고 다음 주에는 딥러닝 자연어 처리를 배울 예정입니다 끝까지 포기하지 마세요'

# 문장을 분리 (sentences)
# 구두점이 존재하지 않더라도 문맥을 파악해서 문장을 리스트로 생성
sentences = kkma.sentences(test_formal)
sentences
```

```python
text = "파이썬에서 자바의 환경 변수 설정은 너무 힘이 듭니다. ㅠㅠ 오프라인 수업이 그립습니닼ㅋㅋ"
kkma.morphs(text)
```

```python
# 명사 추출
print(kkma.nouns(text))
```

#### **sentencepiece**

- 문자의 덩어리를 학습하여 글자 간의 상호작용을 통해서 단어장을 생성한다.
    - 새로운 사전을 만들어주는 것

<br>

- Train()을 이용하여 model을 생성하고 vocab(단어장) 생성
    - `input` : 모델에서 학습할 문자들의 모음 (txt)
    - `model_prefix` : 모델의 이름을 지정
    - `vocab_size`: 단어 사전의 크기 (단어의 개수)
        - 8,000 / 16,000 / 32,000을 일반적으로 사용
    - `model_type`
        - 토큰의 생성 방식
        - `unigram`: BERT, KoGPT 등에서 사용되는 언어 모델 방식 (한 단어씩 잘라서 표현)
        - `bpe`: GPT-2 사용 방식 (한글에서는 부적합)
        - `char`: 문자 단위로 나눠줌 (정보가 너무 잘게 쪼개지는 형태)
        - `word`: 단어 단위로 나눠줌 (한국어에서 부적합)
    - `character_coverage`
        - 학습에 포함할 문자의 종류 비율
        - 1.0을 사용하게 되면 모든 문자를 학습에서 사용
        - 0.9995는 한글, 영문, 숫자 포함
    - `shuffle_input_sentence`
        - 샘플링시 문장 순서를 섞어서 사용할 것인가
        - 모델이 특정 순서에 편향되는 것을 방지
    - `input_sentence_size`
        - 학습 문장을 샘플링
        - 모든 데이터를 사용하는 것이 가장 좋은 방법 (시간이 오래 걸린다)
        - 일부만 샘플링하여 사용하는 방법

<br>

- model load, 서브워드로 문자를 나눠준다.

```python
spm.SentencePieceTrainer.Train(
    input = 'test.txt',
    model_prefix = 'ko_unigram',
    vocab_size = 8000,
    model_type = 'unigram',
    character_coverage = 0.9995,
    input_sentence_size = 100000,
    shuffle_input_sentence = True
)
```

```python
# 모델 로드

sp = spm.SentencePieceProcessor()

sp.load('ko_unigram.model')
```

```python
text4 = '나는 학교에 간다'
text5 = '스토리 개연성이 부족하지만 배우들의 연기는 뛰어났다'
```

```python
# 단어별로 데이터를 나눠주는 작업

print(sp.encode_as_pieces(text4))
print(sp.encode_as_pieces(text5))
```

```python
# 이때 언더바가 키보드에서 사용하는 '_'와는 다르다.

test_text2 = sp.encode_as_pieces(text5)
print(test_text2)

char = '\u2581'
print(char)
```

```python
print('복원:', sp.decode_pieces(test_text2))
```

```python
# 분할된 데이터에서 문자가 아닌 ID를 추출
ids = sp.encode_as_ids(text5)
print(ids)
```

```python
print(sp.decode_ids(ids))
```
