#### **동시 출현 행렬**

- 특정 단어를 기준으로 주변의 범위 안에서 어떠한 단어가 등장했는가?
    - 빈도 수를 확인하는 방법
- 예시
    - '오늘 날씨가 너무 좋다.'
    - '내일 날씨가 조금 흐리다.'
- 범위(window)가 1인 경우
    - 날씨가 → 주변 단어('오늘', '너무', '내일', '조금')
- N-gram 차이
    - N-gram: 단어의 순서 반영
    - 동시출현행렬: 의미 관계를 반영


#### **PMI**
- 두 단어가 우연히 함께 등장했는가? 아니면 의미적으로 연관성이 있는가? 측정
- 측정 값이 클수록 의미적으로 강하게 연결되어있다


#### **PPMI**
- PMI가 음수인 경우는 극히 드물다
- 음수인 데이터를 0으로 대체하는 값

```python
import math
import pandas as pd
from konlpy.tag import Okt
```

```python
docs = [
    '오늘 날씨가 너무 좋다',
    '오늘 기분이 정말 좋다',
    '내일 날씨가 조금 흐리다',
    '기분이 매우 나쁘다'
]
```

```python
# 좌우 단어의 검색 구간을 지정
window_size = 2
```

```python
# 분석기를 이용하여 문장을 형태소 별로 분리
okt = Okt()
tokens = [
    okt.morphs(doc) for doc in docs
]
tokens
```

```python
new_list = []
for token in tokens:
    new_list.extend(token)
new_list
```

```python
list(set(sum(tokens, [])))
```

```python
vocab = sorted(set(sum(tokens, [])))
print(vocab)
```

```python
# 단어들의 목록을 인덱스와 함께 dict 형태로 저장

vocab_dict = {
    word: idx for idx, word in enumerate(vocab)
}

vocab_dict
```

```python
# 동시 출현 행렬을 만들기 위해서 0행렬을 생성
co_matrix = [[0] * len(vocab) for i in range(len(vocab))]
co_matrix
```

```python
tokens
```

```python
# window_size를 기반으로 동시출현 카운트를 체크
for token in tokens:
    for idx, word in enumerate(token):
        count_idx = vocab_dict[word]
        # 윈도우의 범위 지정
        start = max(0, idx - window_size)
        end = min(len(token), idx + window_size)
        # start, end를 이용하여 반복문 생성
        for i in range(start, end):
            if idx != i:
                context = token[i]
                context_idx = vocab_dict[context]
                # co_metric에 특정 위치에 1을 더해준다.
                co_matrix[count_idx][context_idx] += 1
```

```python
co_matrix
```

```python
co_df = pd.DataFrame(co_matrix, index = vocab, columns = vocab)
co_df
```

```python
# PMI 계산
# 각 단어별 동시 등장 횟수의 합계 / total_count

total_count = sum( sum(row) for row in co_matrix )
total_count
```

```python
co_df.sum().sum()
```

```python
p_word = [sum(row) / total_count for row in co_matrix]
p_word
```

```python
p_context = [
    sum(
        co_matrix[i][j] for i in range(len(vocab))
        ) / total_count for j in range(len(vocab))
]

p_context
```

```python
p_context = [
    sum(row) / total_count for row in co_df.T.values
]

p_context
```

```python
def calc_pmi(i, j):
    p_wc = co_matrix[i][j] / total_count
    if p_wc == 0:
        result = 0
    else:
        result = math.log2( p_wc / ( p_word[i] * p_context[j] ) + 1e-12 )
    return result

pmi_matrix = [
                [
                    calc_pmi(i, j) for j in range(len(vocab))
                ]
                for i in range(len(vocab))
            ]


pmi_df = pd.DataFrame(pmi_matrix, index = vocab, columns = vocab)
```
