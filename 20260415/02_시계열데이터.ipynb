### 시계열 데이터

강사님 마크다운 가져오기

```python
# datetime 라이브러리 호출
# datetime 라이브러리 안에 datetime 기능만 가져오겠다

from datetime import datetime
```

```python
time_str = "2026-04-15"

type(time_str)
```

문자열 데이터를 시계열 데이터로 변경

1. strptime(): 문자열 데이터를 시계열 데이터로 변환  
    필수 인자 2개:  
            문자열 데이터  
            해당 문자열의 포맷이 어떻게 되어있는가?  
<br>
2. strftime(): 시계열 데이터를 문자 데이터로 변환
    () 안에는 format 작성

```python
# strptime

time_data = datetime.strptime(time_str, '%Y-%m-%d')
time_data

# 년 월 일 시 분
```

```python
# strftime()

time_data.strftime('%y/%m/%d')
```

```python
time_str2 = '2023년 10월 01일 17시 30분 20초'
# time_str2를 시계열 데이터로 변환
```

```python
time_data2 = datetime.strptime(time_str2, '%Y년 %m월 %d일 %H시 %M분 %S초')
time_data2
```

```python
# 해당하는 날짜의 요일이 언제인가?
time_data2.strftime('%A')
```

pandas에서 시계열 데이터를 이용하는 방법

1. Timestamp() → 시계열 생성 class
    - 1970-01-01을 기준으로 int형 데이터가 들어오면 나노초만큼 증가 

```python
import pandas as pd

pd.Timestamp(1000000)
```

```python
# 날짜 계산

pd.Timestamp(100, unit='D')     # 100일 뒤
```

```python
pd.Timestamp('2026-04-15')
```

```python
pd.Timestamp('10001010091020')
```

```python
# pandas, datetime의 시계열 데이터의 차이는 자동화
# pandas는 일반적인 날짜의 포맷을 저장 (단, 한글형 포맷은 없다)
# to_ 함수() → 데이터의 타입을 변경하는 함수
    # to_type(): 해당 타입으로 변경하겠다.

pd.to_datetime('2026-04-15')
```

```python
pd.to_datetime('04/10/26')
```

```python
pd.to_datetime('2026년4월15일', format='%Y년%m월%d일')
```

```python
# 오늘로부터 15일 뒤
# 현재 날짜에서 15일 뒤는 언제인가?

now_date = pd.to_datetime('2026-04-15')
```

```python
now_date + pd.Timedelta(15, unit='D')
```

```python
# 시계열 데이터 - 시계열 데이터?
pd.to_datetime('2026-04-15') - pd.to_datetime('2026-01-01')
```
