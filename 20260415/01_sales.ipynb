#### 연습

1. 원본 데이터 복사본 생성 (복사본 이름 df)
2. 'Order ID', 'Total Revenue', 'Total Cost', 'Total Profit', 'Order Priority' column 제거
3. column의 이름을 변경: 대륙, 국가, 상품종류, 판매채널, 주문일자, 발송일자, 판매개수, 단가, 원가
4. 대륙, 국가를 기준으로 오름차순 정렬
5. 인덱스를 초기화하고 기존의 인덱스는 제거
6. 파생변수 생성
    - 총단가 = 단가 * 판매개수
    - 총원가 = 원가 * 판매개수
    - 총이윤 = 총단가 - 총원가

```python
# 1

df = sales.copy()
```

```python
# 2

df.drop(columns=['Order ID', 'Total Revenue', 'Total Cost', 'Total Profit', 'Order Priority'], inplace = True)
```

```python
# 3

df.columns = ['대륙', '국가', '상품종류', '판매채널', '주문일자', '발송일자', '판매개수', '단가', '원가']
df.head()
```

```python
# 4

df.sort_values(['대륙', '국가'], ascending = True, inplace = True)
```

```python
# 5

df.reset_index(drop=True, inplace=True)
```

```python
# 6

df['총단가'] = df['단가'] * df['판매개수']
df['총원가'] = df['원가'] * df['판매개수']
df['총이윤'] = df['총단가'] - df['총원가']
```

```python
# pivot → table의 data들을 재구조화 한다.
# pandas에서 pivot_table()
# pandas 안에 존재, DF 안에서도 존재
# 모듈 안에 있는 pivot_table()
# DataFrame class 안에 있는 pivot_table()
# 매개변수 index: 어떤 값을 기준으로 테이블 데이터를 배치할 것인가?
    # 인덱스로 사용할 !!!!!!!!!!!!!깃헙보기!!!!!!!!!!!!
# 매개변수 columns: 어떤 값을 기준으로 테이블 데이터를 배치할 것인가?
    # 컬럼으로 사용할
# 매개변수 values: index, columns를 기준으로 데이터를 결합하여 사용할 value는 무엇인가?
# 매개변수 aggfunc: values에 어떤 연산을 할 것인가?
```

```python
# DF class 안에 존재하는 pivot_table() 함수를 사용

df.pivot_table(
    index = '대륙',        # 대륙을 기준으로 그룹화
    values = '총이윤',      # 값으로 쓰고 싶은 부분
    aggfunc = 'sum'
)
```

```python
df.pivot_table(
    index = '대륙',
    columns = '판매채널',
    values = '총이윤',
    aggfunc = ['sum', 'mean']       # 함수를 여러개 잡은 것처럼 i/c/v 모두 여러개 잡을 수 있음
)
```

```python
# 라이브러리(pandas) 안에 pivot_table() 함수 호출
# 모듈 안에서 함수 호출 시 변경할 데이터를 인자로 입력

pd.pivot_table(
    df,
    index = '대륙',
    columns = '판매채널',
    values = '총이윤',
    aggfunc= 'sum'
)
```
