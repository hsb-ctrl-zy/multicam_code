#### **<font color=#FE6B00>변수</font>: 데이터가 저장되는 공간에 주소 대신 사용할 별칭**

- 불변 데이터
    - 문자, 숫자, 튜플
    - 각각의 다른 공간에 데이터가 대입
    - a = 10  / b = a
        - a = 10, b = 10
        - b에 1을 더해도 a는 10 그대로 유지
    - 함수에서 매개변수의 기본값으로 불변 데이터를 사용하면
        - 매번 데이터가 다른 값들이 대입
<br><br>

- 가변 데이터
    - 리스트, 딕셔너리
    - copy() 함수를 이용하지 않으면 같은 위치를 지정 (공유)
    - list_a = [1, 2, 3] / list_b = list_a
        - list_b에 원소를 추가하면 list_a에도 원소 추가
    - 함수에서 매개변수의 기본값으로 가변 데이터를 사용하면
        - 값을 변경하지 않으면 해당 데이터는 유지하는 성향  
        `def func(a, b = [1, 2, 3]): b.append(a) return b`  
        `func(1) → [1, 2, 3, 1]`
            - 이 때 func 안에 b라는 지역변수는 [1, 2, 3, 1]을 유지
            - 한번 더 실행한다면 [1, 2, 3, 1, 1] 출력
        - copy() 함수 이용 권장

#### <font color=#FE6B00>**반복문**</font>

- **for**
    - for 변수명 in range(시작값, 종료값, 증가값)
        - 시작값부터 종료값-1 까지 증가값만큼 값을 변화시키면서 반복 실행
    - for 변수명 in 자료형 데이터(list, dict, tuple)
        - 자료형 데이터의 각 원소들을 대입하면서 반복 실행
        - 2차원 데이터가 들어오면 `len(자료형데이터)`만큼 반복
            - ex) [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
                - 해당 경우 3번 반복: 1회-[1, 2, 3], 2회-[4, 5, 6], 3회-[7, 8, 9]
<br>

- **while**
    - 조건식이 거짓이 될 때까지 반복 실행하는 반복문
        - 조건이 거짓이 되지 않으면 무한 반복
    - 조건이 거짓이 될 수 있도록 특정 값을 변화 (ex. i += 1)
    - 반복 횟수가 명확하지 않을 때 for문보다 적합할 수 있음
    - **break**: 반복문을 강제로 종료
    - **continue**: 밑 코드 무시하고 반복문으로 되돌아감.

```python
# 어제의 문제
# 클래스 변수가 불변 데이터인 경우: 객체들끼리 공유를 시키는 법

class Class_1:
    text = 'Hello'

    def __init__(self, a, b):
        self.a = a
        self.b = b
        self.text = Class_1.text
        # 불변 데이터의 = 의 의미는 다른 공간에 데이터를 저장 (독립 데이터 생성)
```

```python
class1 = Class_1(10, 20)
class2 = Class_1(50, 60)
```

```python
class1.text = 'Hi'
```

```python
class2.text

# 클래스 변수를 불러왔을 때 독립적인 공간을 만들어버린 것.
```

```python
# 리스트의 경우 정상작동

class Class_2:
    list_a = []

    def __init__(self, a):
        self.a = a
        self.b = Class_2.list_a
        # 가변 데이터에서 =은 얕은 복사 → 같은 주소를 공유한다.
```

```python
class3 = Class_2(10)
class4 = Class_2(30)
```

```python
# 클래스 변수에 50을 추가
class3.b.append(50)
```

```python
print(class3.b, class4.b)
```

```python
# 불변 데이터가 클래스 변수일 때 공유를 시켜보자

class Class_3:
    text = 'Hello'

    def __init__(self, a):
        self.a = a
    
    # 클래스 변수의 데이터를 변경하는 함수

    def func_1(self, _text):
        Class_3.text = _text
        # 접근 방식이 Class_3 안에 text에 데이터를 변경
    
```

```python
class5 = Class_3(10)
class6 = Class_3(40)
```

```python
print(class5.text, class6.text)
```

```python
class5.func_1('Hi')
```

```python
print(class5.text, class6.text)
```

```python
# gemini

class Test:
    text = 'Hello'

    # 일반적인 class 내부의 함수에서 첫번째 매개변수에는
        # 생성이 된 class의 주소값이 자동적으로 입력
    def __init__(self, a):
        self.a = a
    
    # class 이름에 접근할 수 있는 데코레이터
    # 데코레이터가 들어가 있는 함수의 첫번째 매개변수에는
        # 클래스의 이름이 자동적으로 입력
    @classmethod
    def func_1(self, _text):
        # classmethod를 이용한 함수는 첫번째 매개변수명을 cls라고 적음
        # cls에는 해당 class의 이름이 인자로 사용
        # 사용 이유: 수정 중에 class 이름이 변하더라도 cls는 그대로 유지 가능
        cls.text = _text
```

```python
class7 = Test(10)
class8 = Test(50)
```

```python
print(class7.text, class8.text)
```

```python
# 고친 버전

class Test:
    text = 'Hello'

    # 일반적인 class 내부의 함수에서 첫번째 매개변수에는
        # 생성이 된 class의 주소값이 자동적으로 입력
    def __init__(self, a):
        self.a = a
    
    # class 이름에 접근할 수 있는 데코레이터
    # 데코레이터가 들어가 있는 함수의 첫번째 매개변수에는
        # 클래스의 이름이 자동적으로 입력
    @classmethod
    def func_1(cls, _text):
        # classmethod를 이용한 함수는 첫번째 매개변수명을 cls라고 적음
        # cls에는 해당 class의 이름이 인자로 사용
        # 사용 이유: 수정 중에 class 이름이 변하더라도 cls는 그대로 유지 가능
        cls.text = _text
```

```python
class7 = Test(10)
class8 = Test(50)
```

```python
print(class7.text, class8.text)
```
