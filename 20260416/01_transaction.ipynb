### 강사님 풀이

```python
item_master = pd.read_csv('../csv/item_master.csv')
customer_master = pd.read_csv('../csv/customer_master.csv')
```

```python
# 1)
# df에서 작업을 하는데도 inplace=True가 없다

df = transaction.merge(
    item_master,
    on='item_id',
    how='inner',
)
```

```python
df.info()
```

```python
# 2)
# how에 따라서 index 값이 많이 달라짐

customer_master.info()
```

```python
pd.merge(
    df, customer_master,
    on = 'customer_id',
    how = 'right'
).info()
```

```python
df = pd.merge(
    df, customer_master,
    on = 'customer_id',
    how = 'inner'
)
```

```python
list_1 = transaction['customer_id'].unique()
list_2 = customer_master['customer_id'].unique()
```

```python
new_list = []

for customer in list_1:
    if customer not in list_2:
        # 고객의 ID가 list_2에 존재하지 않는다면
        new_list.append(customer)

new_list
```

```python
[customer for customer in list_1 if customer not in list_2]
```

```python
len(set(transaction['customer_id']) - set(customer_master['customer_id']))
```

```python
# 3)
# 문자를 시계열로 변경 (datetime.strptime() // to_datetime())

df['payment_date']
```

```python
df['payment_date'] = pd.to_datetime(df['payment_date'])
df.info()
# dtype: datetime64
```

```python
# strptime() 함수를 이용한다면
# strptime 함수는 Series 형태 변환 불가
    # Series 데이터에서 value들을 하나씩 뽑아서 함수에 넣는다. (map())

from datetime import datetime
```

```python
df['payment_date'].map(
    lambda x: datetime.strptime(x, '%Y-%m-%d %H:%M:%S')
)
```

### 다른 과정

```python
# 요일별 판매가 된 제품들의 총 판매 개수의 합계가 어떻게 되는가?
# 요일이라는 새로운 변수를 생성 : 시계열에서 요일만 따로 추출

df['payment_date'].dt.strftime('%w')
# 0: 일요일

# 다른 방법
# df['payment_date'].dt.weekday
# 0: 월요일
```

```python
df['요일'] = df['payment_date'].dt.strftime('%A')
```

```python
df.head(1)
```

```python
# 요일, item_id를 기준으로 그룹화: quantity 합산

df[['요일', 'item_id', 'quantity']].groupby(['요일', 'item_id']).sum()
```
