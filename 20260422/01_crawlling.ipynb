### 강사님 풀이2

```python
res = requests.get('http://moons-86.iptime.org:8080/')
soup = bs(res.text, 'html.parser')
```

```python
# 접근 방식 2 → 상품명, 카테고리, 가격, 평점, 하이퍼링크를
# 각각의 class의 값들로 접근하여 df 생성
# product-list class를 가진 div를 추출
    # (모든 상품의 정보가 담겨있는 div 영역을 선택)

div_data = soup.find(
    'div',
    attrs = {
        'class': 'product-list'
    }
)
```

```python
# 상품명에 접근: class의 값이 product-title인 태그에 접근

title_list = div_data.find_all(
    attrs = {
        'class' : 'product-title'
    }
)
```

```python
# 어느 경우에 get_text() 를 써도 되는지 확인하기 위해 독자적으로 수행
type(title_list)
```

```python
[title.get_text() for title in title_list]
```

```python
# a 태그에서 href라는 속성의 값을 추출하려면? → Tag['속성명']

link_list = div_data.find_all(
    attrs= {
        'class': 'product-link'
    }
)
```

```python
links = []
for link in link_list:
    # print('http://moons-86.iptime.org:8080/' + link['href'])
    links.append('http://moons-86.iptime.org:8080/' + link['href'])
```

```python
links
```

```python
class_list = ['product-title', 'product-category', 'product-price', 'product-rating', 'product-link']
base_url = 'http://moons-86.iptime.org:8080'
dict_data = {}

# 반복 실행하면서 빈 딕셔너리에 key:[] 값들을 추가
for cls in class_list:
    _list = div_data.find_all(
        attrs = {
            'class': cls
        }
    )

    # cls가 product-link라면 href 속성의 값들을 리스트로 생성
    if cls == 'product-link':
        value = [base_url + link['href'] for link in _list]
    else:
        value = [data.get_text().split(': ')[-1] for data in _list]
    
    # key 값은 cls에서 -로 잘라주고 뒤의 값들을 사용
    key_data = cls.split('-')[-1]
    dict_data[key_data] = value

dict_data
```

```python
df2 = pd.DataFrame(dict_data)
df2
```

```python
# csv 파일로 저장 시 기본값으로 인덱스가 포함되어 저장
    # 인덱스가 의미를 가지고 있을 수 있기 때문

df2.to_csv('test.csv')
```
