### list

```python
# 리스트형 하나 생성

list_a = [10, 20, 30, 40]
list_b = ['a', 'b', 'c']
```

```python
# 리스트의 확장 (두 개의 리스트를 결합)
# tuple에서 +를 이용하면 두 개의 tuple이 결합
    # 리스트에서도 가능한가?

print(list_a + list_b)
```

```python
# 리스트 내장함수 중 확장: extend() 함수가 존재
list_a.extend(list_b)
```

```python
# list_a.extend(list_b) 의 결과값이 출력되지 않음 : 저장이 되었다는 뜻
# 단순히 +를 한 것은 변수에 해당 리스트를 할당하여야 함
print(list_a)

# list_a.extend(list_b) == (list_a = list_a + list_b)
```

```python
# 다차원 리스트 생성 : index(위치(기준))가 여러 개인 데이터

list_c = [
    [10, 20, 30],
    [40, 50, 60]
]   # 2차원 리스트
    # 가독성을 위한 배치

print(list_c)
```

```python
# list_c의 길이는 몇인가? → 2? 6?
    # 2개

print(len(list_c))

# list_c의 기준에서는 [10, 20, 30]이 첫 번째 원소,
# [40, 50, 60]이 두 번째 원소이므로 len(list_c) == 2
```

```python
# 2차원 리스트 데이터에서 단일 데이터로 접근
    # 2번의 위치 선택이 필요 (index 2번 호출)
# ex) list_c 에서 20이라는 원소에 접근
    # 가장 외곽에 있는 괄호부터 차근차근 들어간다고 생각

print("2차원 데이터:", list_c)       # 2차원 데이터
print("1차원 데이터:", list_c[0])    # 1차원 데이터
print("단일 데이터:", list_c[0][1]) # 단일 데이터
```

```python
# 자료형데이터[index] → index가 범위형 데이터([:])가 아니라 단일 데이터[int]라면 차원 -1

print(list_c[:1])       # 대괄호가 그대로 2개: 차원이 유지되고 있는 것
print(list_c[0])        # 대괄호가 1개
```

### dictionary

```python
# dict 형태의 데이터 생성
# dict → 사전형 데이터 → key:value의 쌍으로 데이터가 생성

dict_a = {
    'name' : 'test',
    'age' : 20,
    'phone' : '01012345678'
}

dict_b = {
    'age' : 30,
    'name' : 'test2',
    'phone' : '01011112222'
}
```

```python
# dict 데이터는 위치를 기준으로 데이터를 추출하지 않고 key 값을 기준으로 추출

print(dict_a['name'])
print(dict_b['name'])
```

```python
# CRUD 中 C(Create)
# dict에 데이터 추가 → 새로운 key와 value 사용

dict_a['address'] = 'seoul'
print(dict_a)

# index 대신 key를 사용한다고 봐도 무방
# index와 달리 key는 순서에 영향이 없음
```

```python
# CRUD 中 U(Update)
# 수정 → 기존의 데이터(dict[key])를 제거하고 새로운 데이터(value)를 대입(=)

print(dict_a['age'])
dict_a['age'] = 25
print(dict_a)
```

```python
# CRUD 中 D(Delete)
# 제거 → del : key와 value가 한번에 삭제

del dict_a['phone']
print(dict_a)
```

### 다차원데이터: 복합

```python
# DB나 open api에서 데이터를 불러오는 경우
# 리스트와 딕셔너리가 혼합되어있는 다차원 데이터의 형태

# 1. 딕셔너리 안에 리스트

sample_data = {
    'name' : ['test', 'test2', 'test3'],
    'age' : [20, 30, 40],
    'address' : ['Seoul', 'Busan', 'Gwangju']
}
```

```python
# sample_data에서 test2만 추출하려면?

print(sample_data['name'][1])
```

```python
# sample_data에서 test2라는 사람의 정보를 모두 출력

# ↓ 내가 해본 것: False 출력
    # print(sample_data[['name']]=='test2')
# ↓ 내가 해본 것2: 에러    
    print(sample_data[['name']]=='test2')
```

```python
# 반복문을 이용
# for문 (자료형 데이터에서 각각의 원소들을 추출하여 변수에 대입하여 반복)

for i in sample_data:
    #print(i)
        #i에 뭐가 들어가는 지 확인: key값이 출력
    #print(sample_data[i])
        #value값 확인
    print(sample_data[i][1])
```

```python
# while문 (특정 조건식이 거짓이 될 때까지 반복)
# 시작값 설정: dictionary의 길이가 필요
    # sample_data key 값들의 목록이 필요
        #sample_data.keys()

keys = list(sample_data.keys())

# 초기값 설정
i = 0

# while문 생성
# while i < 3:
    # 3의 의미: sample_data의 길이
while i < len(sample_data):
    # 동적 할당: sample_data의 길이가 변하더라도 사용 가능

    print(keys[i])
        # key 값 출력
        # i 값의 의미는: key 값들의 인덱스
    print(sample_data[keys[i]][1])
        # 해당 인덱스의 2번째 값 출력
        
    print('-'*5)

    i += 1
        # i 값을 1씩 증가 (무한루프 방지)
```

```python
# list에서 sort() 함수: list 오름차순 정렬
# 코딩 테스트 algorithm basic
# sort() 함수 없이 오름차순 정렬

# ---------------------------------------------
# 내가 한 것

sample_list2 = [30, 10, 50, 7]
result_list = []
initial = max(sample_list2)+1
# print(initial)

for i in sample_list2:
    if i < initial:
        initial = i
    # print(initial)
    result_list.append(i)
    sample_list2.remove(i)

print(result_list)
```

```python
# 강사님이 한 것

sample_list = [30, 10, 50, 7]
sort_list = []

# sample_list2의 길이만큼 반복 실행하는 반복문을 생성

for i in range(len(sample_list)):
    
    min_value = sample_list[0]
    # 기준이 되는 list의 첫번째 원소를 변수에 저장

    for j in range(1, len(sample_list), 1):
    # 2번째 데이터부터 min_value와 비교
    # j는 sample_list의 인덱스 (두번째 데이터
        
        if min_value > sample_list[j]:
            min_value = sample_list[j]
            # sample_list[j]가 더 작으면
            # min_value를 sample_list[j]로 수정

    # print(min_value)
    # break
    sort_list.append(min_value)
    sample_list.remove(min_value)
    print(sample_list)
    print(sort_list)

print('-'*30)
print('최종:',sort_list)
```

```python
# 더 쉬운 버전 / min() 활용

sample_list = [30, 10, 50, 7]
sort_list = []

# min() : 여러 개의 데이터 중 가장 작은 값을 반환
# sample_list 에서 가장 작은 값을 추출
    # → sort_list 에 추가
        # → sample_list에서 제거
# sample_list의 길이만큼 반복 실행

for i in range(len(sample_list)):

    min_value = min(sample_list)
    # min() 함수를 이용하여 sample_list 에서 가장 작은 데이터 추출
        # → 변수에 저장
    # print(min_value)
    
    sort_list.append(min_value)
    # sort_list 에 추가

    sample_list.remove(min_value)
    # sample_list 에서 제거

print(sort_list)
```

```python
# 내림차순 정렬
# max() : 여러 개의 데이터 중 가장 큰 값을 반환

sample_list = [30, 10, 50, 7]
sort_list = []

for i in range(len(sample_list)):
    max_value = max(sample_list)    
    sort_list.append(max_value)
    sample_list.remove(max_value)

print(sort_list)
```
