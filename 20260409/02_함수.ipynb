### 문자열 내장 함수

- upper() : 문자열을 모두 대문자로 변환
- lower() : 문자열을 모두 소문자로 변환
- swapcase() : 대문자를 소문자로, 소문자를 대문자로 변환
- title() : 단어의 첫 번째 글자만 대문자로 변환

- strip() : 기본적으로는 문자열의 시작과 끝에 존재하는 공백을 제거
    - lstrip() : 문자열 시작의 공백을 제거
    - rstrip() : 문자열 끝의 공백을 제거

- replace() : 특정한 문자를 찾아서 해당 문자를 다른 문자로 치환
    - 문자 사이에 특정 문자를 치환 
        - strip() 함수는 문자의 시작과 끝만 확인
        - replace() 는 문자를 모두 확인
    
- split() : 특정 문자를 찾아서 문자를 나눠주는 기능
    - 입력값은 문자, 출력값은 list

```python
# document 사이트 (특정 라이브러리들이 운영하는 사이트)
# 함수의 목록
    # 함수명
        # 매개변수명 (required) → 필수 항목 → 기본값이 지정되지 않았다.
            # 또는 no_default
        # 매개변수명 (option) → 선택 항목 → 기본값이 지정되어 있다.


# 인자의 개수가 가변인 경우 (매개변수 앞에 * 붙여준다)
    # 인자에 * : list의 주소를 참조하는가? → No
        # *매개변수 → 인자들은 하나로 묶어준다. (패킹)
        # *인자 → 묶여 있는 데이터를 풀어준다. (언패킹)
            # list의 주소를 참조하는 것이 아니라 리스트 내부 원소의 주소를 참조
                # func(*[1,2,3]) → func(list[0], list[1], list[2])
                # a = list[0], b = list[1], c = list[2]

def func_8(*args):
    # print(args)
    # print(type(args))
    result = 0
    for i in args:
        result += i
    return result

func_8(1,2,3,4)
```

```python
# 인자 개수를 가변으로 받는 매개변수 뒤에 매개변수를 추가로 생성한다면?

def func_9(*args, x, y):
    print(args)
    print(x)
    print(y)

func_9(1,2,3,4,5)
```

```python
# sort() → 오름차순 / 내림차순 정렬을 할 수 있는 새로운 함수
# 오름차순 정렬 → min()
# 내림차순 정렬 → max()

# 함수 선언
    # 매개변수 2개
        # values 인자의 개수가 가변인 형태
        # reverse 기본값이 False(오름차순 정렬) // True(내림차순 정렬)

def custom_sort(*values, reverse=False):
    result = []
        # 정렬이 되는 데이터들을 저장할 공간 (빈 리스트)
    values = list(values)
        # values의 데이터가 수정을 할 수 있도록 list로 변환
    for i in range(len(values)):
            # values의 길이만큼 반복하는 반복문 생성
            # values의 type이 tuple이기 때문에 가능
        if reverse:
            sort_value = max(values)
        else:
            sort_value = min(values)
        result.append(sort_value)
        values.remove(sort_value)
    return result
```

```python
custom_sort(33,19,102,39,reverse=True)
```
