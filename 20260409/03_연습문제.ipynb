### 연습 문제
- 매개변수 2개
    - datas: 인자의 개수가 가변인 매개변수
    - types: 기본값의 설정이 ('max')

- types이 max라면 datas 중 가장 큰 값을 되돌려준다
    - min이라면 datas 중 가장 작은 값을 되돌려준다
    - mean이라면 datas의 평균 값을 되돌려준다
    - types가 위의 3가지 경우가 아니라면 'types 데이터 입력값이 잘못되었습니다' 라는 출력

```python
def basic_stats(*datas, types='max'):
    first_value = datas[0]    
    if types =='max':
        for i in range(1, len(datas), 1):
            if first_value < datas[i]:
                first_value = datas[i]
        return first_value
    
    elif types =='min':
        for i in range(1, len(datas), 1):
            if first_value > datas[i]:
                first_value = datas[i]
        return first_value

    elif types == 'mean':
        a = 0
        for data in datas:
            a += data
        a /= len(datas)
        return round(a, 2)

    else:
        return 'types 데이터 입력값이 잘못되었습니다.'
```

```python
print(basic_stats(3, 1, 9, 93, 29, 10, types='min'))
print(basic_stats(3, 1, 9, 29, 10))
print(basic_stats(3, 1, 9, 93, 29, 10, types='max'))
print(basic_stats(3, 1, 9, 93, 29, 10, types='mean'))
print(basic_stats(3, 1, 9, 93, 29, 10, types='man'))
```

```python
# markdown 부분, 의도적 에러 발생 함수 생성

def func_14(*datas, types='max'):
    if types == 'max':
        result = max(datas)
    
    elif types == 'min':
        result = min(datas)
    
    elif types == 'mean':
        res_sum = sum(datas)
        result = res_sum / len(datas)
    
    else:
        raise ValueError("types 데이터 입력값이 잘못되었습니다.")

    return result
```
